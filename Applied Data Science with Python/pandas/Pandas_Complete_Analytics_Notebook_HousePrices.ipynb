# 9. Text Data in Pandas

Common operations include:

```python
# Example 1: lower-case city names
df["city_lower"] = df["city"].str.lower()
df[["city", "city_lower"]].head()
```

# 12. Joining, Merging, and Concatenating

These operations combine data from multiple tables.

Common tools:
- `concat()`
- `merge()`
- `join()`


```python
# Example 1: concat two small DataFrames
sales_jan = pd.DataFrame({"Month": ["Jan"], "Sales": [1200]})
sales_feb = pd.DataFrame({"Month": ["Feb"], "Sales": [1500]})

combined_sales = pd.concat([sales_jan, sales_feb], ignore_index=True)
combined_sales
```

```python
# Example 2: merge customer and city data
city_info = pd.DataFrame({
    "city": ["Seattle", "Kent", "Shoreline"],
    "region": ["West", "East", "North"]
})

merged = df[["city", "price"]].merge(city_info, on="city", how="left")
merged.head()
```

```python
# Example 3: join using index
left = pd.DataFrame({"A": [1, 2, 3]}, index=["x", "y", "z"])
right = pd.DataFrame({"B": [10, 20, 30]}, index=["x", "y", "z"])
left.join(right)
```

```python
# Example 4: inner merge
example_left = pd.DataFrame({"id": [1, 2, 3], "value_left": [10, 20, 30]})
example_right = pd.DataFrame({"id": [2, 3, 4], "value_right": [200, 300, 400]})
example_left.merge(example_right, on="id", how="inner")
```

```python
# Example 5: outer merge
example_left.merge(example_right, on="id", how="outer")
```

# 14. Reshaping Data

Reshaping changes the structure of a DataFrame without changing the data itself.

Important reshaping techniques:
- `pivot_table()`
- `pivot()`
- `melt()`
- `stack()`
- `unstack()`


```python
# Example 1: pivot table of average price by city and bedrooms
pivot_city_bed = pd.pivot_table(df, values="price", index="city", columns="bedrooms", aggfunc="mean")
pivot_city_bed.head()
```

```python
# Example 2: pivot table of price by price band and waterfront
pd.pivot_table(df, values="price", index="price_band", columns="waterfront", aggfunc="mean")
```

```python
# Example 3: melt a wide-format table
wide = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar"],
    "Revenue_Online": [1200, 1500, 1800],
    "Revenue_Store": [900, 1100, 1300]
})
melted = wide.melt(id_vars="Month", var_name="Channel", value_name="Revenue")
melted
```

```python
# Example 4: stack and unstack
simple = pd.DataFrame({"A": [1, 2], "B": [3, 4]}, index=["X", "Y"])
print(simple.stack())
print(simple.unstack())
```

```python
# Example 5: reset index after grouping
city_count = df.groupby("city")["price"].count()
city_count.reset_index().head()
```

# Step 1 — Basic Dataset Understanding


```python
# Basic profiling
print("Shape:", df.shape)
print("Missing values per column:")
print(df.isna().sum())
print()
print(df[["price", "bedrooms", "bathrooms", "sqft_living", "city"]].head())
```

# Step 2 — Create Derived Columns


```python
# Derived columns for analysis
df["price_lakhs"] = df["price"] / 100000
df["house_age"] = 2024 - df["yr_built"]
df["is_renovated"] = np.where(df["yr_renovated"] > 0, "Yes", "No")
df["price_band"] = pd.cut(
    df["price"],
    bins=[0, 300000, 600000, 1000000, np.inf],
    labels=["Low", "Mid", "High", "Luxury"],
    include_lowest=True
)

df[["price", "price_lakhs", "house_age", "is_renovated", "price_band"]].head()
```

# Step 3 — City-wise Business Insight


```python
city_summary = df.groupby("city").agg(
    avg_price=("price", "mean"),
    max_price=("price", "max"),
    total_houses=("price", "count"),
    avg_sqft=("sqft_living", "mean")
).sort_values("avg_price", ascending=False)

city_summary.head(10)
```

# Step 4 — Bedrooms and Price Analysis


```python
bedroom_summary = df.groupby("bedrooms").agg(
    avg_price=("price", "mean"),
    count_houses=("price", "count"),
    avg_sqft=("sqft_living", "mean")
).sort_index()

bedroom_summary
```

# Step 5 — Date Insight


```python
df["sale_month"] = df["date"].dt.month
df["sale_weekday"] = df["date"].dt.day_name()

monthly_insight = df.groupby("sale_month")["price"].agg(["count", "mean", "max"])
weekday_insight = df.groupby("sale_weekday")["price"].mean()

print("Monthly insight:")
display(monthly_insight)
print("Weekday insight:")
display(weekday_insight)
```

# Step 6 — Top and Bottom Records
