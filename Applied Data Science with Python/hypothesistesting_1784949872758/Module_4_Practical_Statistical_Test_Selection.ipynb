# Module 4: Statistical Test Selection — A Practical Guide

In the previous module, we used different statistical tests to answer different types of questions.

The challenge in practice is often not running the test.

The challenge is choosing the correct test.

This short module provides a practical decision framework that can be used before starting statistical analysis.

The basic idea is:

$$
\boxed{
\text{Question}
\rightarrow
\text{Variable Type}
\rightarrow
\text{Number of Groups}
\rightarrow
\text{Relationship Between Groups}
\rightarrow
\text{Choose Test}
}
$$

# 1. Start With the Business Question

Before choosing a statistical test, ask:

> What exactly am I trying to compare?

Most problems in this class fall into one of these categories:

1. Compare one sample with a benchmark
2. Compare two independent groups
3. Compare the same group before and after
4. Compare three or more groups
5. Check association between categorical variables
6. Compare variability

Once the question is clear, test selection becomes much easier.

# 2. Quick Test Selection Table

| Business Question | Data Situation | Test |
|---|---|---|
| Is one average different from a benchmark? | One numerical sample vs known value | One-Sample T-Test |
| Is one average different from a benchmark when $\sigma$ is known? | Numerical sample with known population SD | Z-Test |
| Are two separate groups different? | Numerical outcome, two independent groups | Independent T-Test |
| Did the same subjects change? | Numerical outcome, before and after | Paired T-Test |
| Are two categorical variables associated? | Categorical vs categorical | Chi-Square Test |
| Are 3 or more group averages different? | Numerical outcome across 3+ groups | One-Way ANOVA |
| Do two factors affect a numerical outcome? | Two categorical factors, one numerical outcome | Two-Way ANOVA |
| Do two groups have different variability? | Comparing two variances | F-Test |

This table is the main reference for the remaining analysis.

# 3. Decision Roadmap

Use the following sequence.

## Step 1: Is the outcome numerical or categorical?

### Numerical Outcome

Ask how many groups or conditions are being compared.

### Categorical Variables

If the question is about association between two categorical variables:

$$
\rightarrow
\text{Chi-Square Test}
$$

---

## Step 2: For a Numerical Outcome

### One sample vs a benchmark

$$
\rightarrow
\text{One-Sample T-Test}
$$

If the population standard deviation is known:

$$
\rightarrow
\text{Z-Test}
$$

### Two groups

Ask:

> Are the groups independent?

If yes:

$$
\rightarrow
\text{Independent T-Test}
$$

If the same observations are measured twice:

$$
\rightarrow
\text{Paired T-Test}
$$

### Three or more groups

$$
\rightarrow
\text{One-Way ANOVA}
$$

### Two categorical factors

$$
\rightarrow
\text{Two-Way ANOVA}
$$

# 4. Quick Case 1: Delivery Benchmark

A delivery company claims:

$$
\mu=30\text{ minutes}
$$

A sample of delivery times is collected.

Question:

> Is the average delivery time different from 30 minutes?

### Identify the structure

- Outcome: Numerical
- Number of samples: One
- Comparison: Sample mean vs benchmark

### Test

$$
\boxed{\text{One-Sample T-Test}}
$$

```python
from scipy import stats
import numpy as np

delivery_times = np.array([
    31, 32, 30, 34, 33,
    29, 32, 31, 35, 33
])

statistic, p_value = stats.ttest_1samp(
    delivery_times,
    popmean=30
)

print("P-Value:", round(p_value, 4))
```

# 5. Quick Case 2: Campaign A vs Campaign B

A company wants to compare average customer spending between two marketing campaigns.

The customers in Campaign A are different from the customers in Campaign B.

### Identify the structure

- Outcome: Numerical
- Groups: Two
- Relationship: Independent

### Test

$$
\boxed{\text{Independent T-Test}}
$$

```python
campaign_a = [2200, 2400, 2300, 2500, 2350]
campaign_b = [1900, 2000, 1950, 2100, 2050]

statistic, p_value = stats.ttest_ind(
    campaign_a,
    campaign_b,
    equal_var=False
)

print("P-Value:", round(p_value, 4))
```

# 6. Quick Case 3: Before vs After

Employees are measured before and after completing a training program.

### Identify the structure

- Outcome: Numerical
- Measurements: Two
- Relationship: Same employees

### Test

$$
\boxed{\text{Paired T-Test}}
$$

The key word is **same**.

Whenever the same person, product, store, machine, or unit is measured twice, consider whether the observations are paired.

```python
before = [62, 65, 60, 68, 64]
after = [68, 70, 66, 72, 69]

statistic, p_value = stats.ttest_rel(
    after,
    before
)

print("P-Value:", round(p_value, 4))
```

# 7. Quick Case 4: Churn vs Contract Type

A company wants to know whether customer churn is associated with contract type.

Variables:

- Churn: Yes / No
- Contract: Monthly / Annual

### Identify the structure

- Variable 1: Categorical
- Variable 2: Categorical
- Question: Association

### Test

$$
\boxed{\text{Chi-Square Test of Independence}}
$$

```python
import pandas as pd

table = pd.DataFrame(
    [[40, 60],
     [15, 85]],
    index=["Monthly", "Annual"],
    columns=["Churn Yes", "Churn No"]
)

chi2, p_value, dof, expected = stats.chi2_contingency(table)

print("P-Value:", round(p_value, 4))
```

# 8. Quick Case 5: Sales Across Four Regions

A company wants to compare average sales across:

- North
- South
- East
- West

### Identify the structure

- Outcome: Numerical
- Groups: Four
- Groups are independent

### Test

$$
\boxed{\text{One-Way ANOVA}}
$$

A T-Test is designed for comparing two means.

ANOVA extends the comparison to three or more groups.

```python
north = [22, 24, 23, 25]
south = [18, 19, 20, 19]
east = [21, 22, 20, 23]
west = [25, 26, 24, 27]

statistic, p_value = stats.f_oneway(
    north,
    south,
    east,
    west
)

print("P-Value:", round(p_value, 4))
```

# 9. One Decision Pattern for All Tests

Once the correct test is selected, the decision logic remains the same.

Use:

$$
\alpha=0.05
$$

Then:

$$
p<0.05
\Rightarrow
\text{Reject }H_0
$$

Otherwise:

$$
\text{Fail to Reject }H_0
$$

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("The result is statistically significant.")
else:
    print("Fail to Reject H0")
    print("The result is not statistically significant.")
```

Use this sequence:

$$
\boxed{

- Same group before and after → Paired T-Test
- Known population standard deviation → Z-Test
- Categorical vs categorical → Chi-Square Test
- Three or more means → One-Way ANOVA
- Two factors affecting a numerical outcome → Two-Way ANOVA
- Two variances → F-Test
