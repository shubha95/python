# 2. The Core Idea of Hypothesis Testing

Suppose a company claims:

$$
\mu = 30
$$

where $\mu$ represents the true average delivery time.

We collect a sample.

The sample mean is:

$$
\bar{x} = 32
$$

There is clearly a difference:

$$
32 - 30 = 2
$$

But the important question is not:

> Is there a difference?

There is.

The important question is:

> Is the difference large enough that it is unlikely to be explained by random sampling variation alone?

Hypothesis testing helps us answer this question.

# 3. Null and Alternative Hypotheses

The Null Hypothesis is written as:

$$
H_0
$$

It usually represents:

- No difference
- No effect
- No association
- No change

The Alternative Hypothesis is written as:

$$
H_1
$$

It represents the possibility that a meaningful difference, effect, association, or change exists.

For the delivery-time example:

$$
H_0:\mu=30
$$

$$
H_1:\mu\neq30
$$

The sample data will provide evidence either against or in support of retaining the null hypothesis.

# 4. Significance Level

Before performing the test, we choose a significance level.

A common choice is:

$$
\alpha = 0.05
$$

This acts as our decision threshold.

The significance level represents the level of evidence we require before rejecting the Null Hypothesis.

Throughout most examples in this notebook, we will use:

$$
\alpha=0.05
$$

# 5. Understanding the P-Value

The p-value tells us how surprising our observed result would be if the Null Hypothesis were true.

A small p-value means:

> The observed result would be unusual under $H_0$.

This provides evidence against the Null Hypothesis.

Our decision rule is:

$$
p < \alpha
\Rightarrow
\text{Reject }H_0
$$

$$
p \geq \alpha
\Rightarrow
\text{Fail to Reject }H_0
$$

With:

$$
\alpha=0.05
$$

we use:

$$
p<0.05
\Rightarrow
\text{Statistically Significant}
$$

## Important Language

If:

$$
p \geq 0.05
$$

we say:

> Fail to reject the Null Hypothesis.

We do not say:

> Accept the Null Hypothesis as true.

The data may simply not provide enough evidence to reject it.

# 6. One-Tailed and Two-Tailed Questions

The form of the Alternative Hypothesis depends on the business question.

### Two-Tailed Test

Question:

> Is the average delivery time different from 30 minutes?

$$
H_0:\mu=30
$$

$$
H_1:\mu\neq30
$$

---

### Right-Tailed Test

Question:

> Is the average delivery time greater than 30 minutes?

$$
H_0:\mu\leq30
$$

$$
H_1:\mu>30
$$

---

### Left-Tailed Test

Question:

> Has the average delivery time fallen below 30 minutes?

$$
H_0:\mu\geq30
$$

$$
H_1:\mu<30
$$

The direction of the test should come from the original question, not from looking at the data after it has been collected.

# 7. Type I and Type II Errors

Statistical decisions can be wrong.

Consider a fraud detection system.

### Type I Error

A genuine transaction is classified as fraud.

We reject a true Null Hypothesis.

$$
P(\text{Type I Error})=\alpha
$$

---

### Type II Error

A fraudulent transaction is treated as genuine.

We fail to reject a false Null Hypothesis.

The probability of a Type II Error is represented by:

$$
\beta
$$

---

The importance of each error depends on the problem.

In medical diagnosis, fraud detection, manufacturing, credit risk, and predictive analytics, the cost of the two errors may be very different.

Statistical decisions should therefore be interpreted in the context of the real-world consequences.

# Case Study 1: One-Sample T-Test

## Business Problem

A food delivery company claims that its average delivery time is 30 minutes.

A random sample of recent deliveries is collected.

The question is:

> Is the average delivery time significantly different from 30 minutes?

We have:

- One numerical variable: Delivery Time
- One sample
- One benchmark value: 30 minutes

This is a **One-Sample T-Test** problem.

````python
delivery_times = np.array([
    31, 33, 29, 34, 32, 30, 35, 31, 33, 32,
    30, 34, 31, 36, 32, 29, 33, 34, 32, 31,
    35, 30, 33, 32, 34, 31, 33, 35, 30, 32
])

claimed_mean = 30

print("Sample Size:", len(delivery_times))
print("Sample Mean:", round(delivery_times.mean(), 2))
print("Company Claim:", claimed_mean)
````

## Step 1: Define the Hypotheses

The company claim becomes the Null Hypothesis:

$$
H_0:\mu=30
$$

We want to know whether the actual average is different:

$$
H_1:\mu\neq30
$$

This is a two-tailed test.

## Step 2: Perform the Test

We use:

```python
stats.ttest_1samp()
```

The test compares the sample mean with the claimed population mean.

````python
t_statistic, p_value = stats.ttest_1samp(
    delivery_times,
    popmean=claimed_mean
)

print("T-Statistic:", round(t_statistic, 3))
print("P-Value:", round(p_value, 5))
````

## Step 3: Make the Decision

````python
if p_value < alpha:
    print("Reject H0")
    print("The average delivery time is significantly different from 30 minutes.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence to conclude that the average delivery time differs from 30 minutes.")
````

## What Does This Mean?

The test does not ask whether the sample mean is exactly 30.

It asks whether the observed difference between the sample mean and 30 is large relative to the variability in the sample.

This is the key idea behind statistical significance.

### Predictive Analytics Connection

Historical benchmarks are often used when building forecasts and predictive systems.

Before treating a historical benchmark as reliable, we may want to test whether current data still behaves consistently with that benchmark.

# Case Study 2: Independent Two-Sample T-Test

## Business Problem

A marketing team runs two different campaigns.

Campaign A is shown to one group of customers.

Campaign B is shown to another independent group.

The company wants to know:

> Do customers exposed to the two campaigns spend different amounts on average?

We have:

- One numerical outcome: Customer Spending
- Two independent groups: Campaign A and Campaign B

This is an **Independent Two-Sample T-Test** problem.

````python
campaign_a = np.array([
    2200, 2400, 2100, 2500, 2300,
    2600, 2250, 2450, 2350, 2550
])

campaign_b = np.array([
    1800, 1950, 2000, 2100, 1900,
    2050, 1850, 2150, 2000, 1950
])

print("Campaign A Mean:", campaign_a.mean())
print("Campaign B Mean:", campaign_b.mean())
````

## Hypotheses

$$
H_0:\mu_A=\mu_B
$$

$$
H_1:\mu_A\neq\mu_B
$$

The Null Hypothesis says that the average spending is the same for both campaigns.

The Alternative Hypothesis says that the average spending differs.

````python
t_statistic, p_value = stats.ttest_ind(
    campaign_a,
    campaign_b,
    equal_var=False
)

print("T-Statistic:", round(t_statistic, 3))
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("Average customer spending differs significantly between the campaigns.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence of a difference in average spending.")
````

## Practical Interpretation

A significant result suggests that campaign membership is associated with differences in average spending.

This does not automatically prove that the campaign caused the difference.

Other factors may also differ between the groups unless the experiment was properly controlled.

### Predictive Analytics Connection

When a variable separates groups with meaningfully different outcomes, it may contain useful predictive information.

However:

$$
\text{Statistical Significance}
\neq
\text{Guaranteed Predictive Importance}
$$

Predictive usefulness must eventually be validated using a model and unseen data.

# Case Study 3: Paired T-Test

## Business Problem

A company provides a training program to 12 employees.

Productivity is measured:

- Before training
- After training

The question is:

> Did employee productivity change after training?

The two sets of observations come from the same employees.

The observations are therefore paired.

This is a **Paired T-Test** problem.

````python
before_training = np.array([
    62, 65, 60, 70, 68, 64,
    66, 61, 69, 63, 67, 65
])

after_training = np.array([
    68, 70, 66, 74, 72, 69,
    71, 67, 73, 68, 72, 70
])

training_data = pd.DataFrame({
    "Before": before_training,
    "After": after_training
})

training_data["Difference"] = (
    training_data["After"] -
    training_data["Before"]
)

training_data
````

## Hypotheses

For a two-tailed test:

$$
H_0:\mu_{\text{difference}}=0
$$

$$
H_1:\mu_{\text{difference}}\neq0
$$

If the business question specifically asks whether productivity **increased**, a one-tailed alternative could be defined before analysing the data.

````python
t_statistic, p_value = stats.ttest_rel(
    after_training,
    before_training
)

print("T-Statistic:", round(t_statistic, 3))
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("Employee productivity changed significantly after training.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence of a change in productivity.")
````

## Why Not Use an Independent T-Test?

The same employees appear in both groups.

The observation:

$$
\text{Employee 1 Before}
$$

is directly related to:

$$
\text{Employee 1 After}
$$

The test should therefore analyse the differences within each pair.

### Predictive Analytics Connection

Paired comparisons are common when evaluating interventions:

- Before and after training
- Before and after treatment
- Before and after a product change
- Before and after a policy change

# Case Study 4: Z-Test

## Business Problem

Suppose a company historically knows that customer order values have:

$$
\mu=₹2000
$$

and population standard deviation:

$$
\sigma=₹400
$$

A large sample of recent customers has an average order value of ₹2080.

The company wants to know:

> Has the average order value changed significantly?

When the population standard deviation is known, a Z-Test can be used.

The Z-statistic is:

$$
Z=
\frac{\bar{x}-\mu}
{\sigma/\sqrt{n}}
$$

````python
sample_mean = 2080
population_mean = 2000
population_std = 400
n = 100

z_statistic = (
    sample_mean - population_mean
) / (
    population_std / np.sqrt(n)
)

p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))

print("Z-Statistic:", round(z_statistic, 3))
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("The average order value is significantly different from ₹2000.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence that average order value has changed.")
````

## Z-Test or T-Test?

A practical rule is:

### Use a Z-Test when:

- The population standard deviation $\sigma$ is known
- Often the sample is reasonably large

### Use a T-Test when:

- The population standard deviation is unknown
- The sample standard deviation $s$ is used instead

In most real-world datasets, the population standard deviation is unknown.

For this reason, T-Tests are encountered more frequently in practical analysis.

# Case Study 5: Chi-Square Test of Independence

## Business Problem

A subscription company wants to understand customer churn.

Two variables are available:

### Contract Type

- Monthly
- Annual

### Churn

- Yes
- No

The question is:

> Is customer churn associated with contract type?

Both variables are categorical.

This is a **Chi-Square Test of Independence** problem.

````python
churn_data = pd.DataFrame({
    "Contract": (
        ["Monthly"] * 100 +
        ["Annual"] * 100
    ),
    "Churn": (
        ["Yes"] * 40 + ["No"] * 60 +
        ["Yes"] * 15 + ["No"] * 85
    )
})

contingency_table = pd.crosstab(
    churn_data["Contract"],
    churn_data["Churn"]
)

contingency_table
````

## Hypotheses

$$
H_0:
\text{Contract Type and Churn are independent}
$$

$$
H_1:
\text{Contract Type and Churn are associated}
$$

````python
chi2_statistic, p_value, dof, expected = stats.chi2_contingency(
    contingency_table
)

print("Chi-Square Statistic:", round(chi2_statistic, 3))
print("Degrees of Freedom:", dof)
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("Contract type and churn are significantly associated.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence of an association between contract type and churn.")
````

## Practical Interpretation

If the result is significant, churn behaviour differs across contract categories.

### Predictive Analytics Connection

This is especially useful when exploring categorical predictors for a categorical target.

For example:

$$
\text{Contract Type}
\rightarrow
\text{Churn}
$$

A statistically significant association suggests that the predictor may contain information related to the target.

Again, this is evidence of association, not proof of causation or guaranteed model importance.

## Practice Problem

A bank wants to know whether loan default is associated with employment type.

Variables:

- Employment Type: Salaried, Self-Employed
- Default: Yes, No

Create a contingency table and perform a Chi-Square Test of Independence.

# Case Study 6: One-Way ANOVA

## Business Problem

A retail company operates in four regions:

- North
- South
- East
- West

The company wants to know:

> Does average customer spending differ across the four regions?

The outcome is numerical.

There are more than two independent groups.

This is a **One-Way ANOVA** problem.

````python
north = np.array([2200, 2300, 2250, 2400, 2350, 2280])
south = np.array([1800, 1900, 1850, 1950, 1880, 1920])
east  = np.array([2100, 2150, 2200, 2050, 2180, 2120])
west  = np.array([2500, 2450, 2550, 2600, 2480, 2520])

print("North Mean:", round(north.mean(), 2))
print("South Mean:", round(south.mean(), 2))
print("East Mean:", round(east.mean(), 2))
print("West Mean:", round(west.mean(), 2))
````

## Hypotheses

$$
H_0:
\mu_{\text{North}}
=
\mu_{\text{South}}
=
\mu_{\text{East}}
=
\mu_{\text{West}}
$$

The Alternative Hypothesis is:

$$
H_1:
\text{At least one group mean is different}
$$

ANOVA does not initially tell us which specific groups differ.

It tests whether there is evidence that at least one mean differs from the others.

````python
f_statistic, p_value = stats.f_oneway(
    north,
    south,
    east,
    west
)

print("F-Statistic:", round(f_statistic, 3))
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("Average customer spending differs across at least one region.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence of a difference in average spending across regions.")
````

## Why Not Run Many T-Tests?

If we repeatedly perform separate T-Tests between every pair of groups, the probability of making a Type I Error increases.

ANOVA provides a single overall test for comparing multiple means.

If ANOVA is significant, post-hoc tests can then be used to identify which groups differ.

### Predictive Analytics Connection

Suppose `Region` is a potential predictor and `Spending` is the target.

ANOVA helps us investigate whether average target values vary across categories of the predictor.

# Case Study 7: Two-Way ANOVA

## Business Problem

Suppose crop yield may depend on two factors:

- Fertilizer Type
- Water Level

The target is:

- Crop Yield

Now we have two categorical factors affecting one numerical outcome.

A Two-Way ANOVA can investigate:

1. Does fertilizer type affect yield?
2. Does water level affect yield?
3. Does the effect of fertilizer depend on water level?

The third question is called an **interaction effect**.

This is important because real-world variables often work together rather than independently.

````python
import statsmodels.api as sm
from statsmodels.formula.api import ols

crop_data = pd.DataFrame({
    "Fertilizer": ["A","A","A","A","B","B","B","B"] * 2,
    "Water": (
        ["Low"] * 8 +
        ["High"] * 8
    ),
    "Yield": [
        40,42,41,43,48,49,47,50,
        55,57,56,58,70,72,69,71
    ]
})

model = ols(
    "Yield ~ C(Fertilizer) * C(Water)",
    data=crop_data
).fit()

anova_table = sm.stats.anova_lm(
    model,
    typ=2
)

anova_table
````

## Reading the Two-Way ANOVA Table

Look at the p-values for:

### Fertilizer

Tests whether fertilizer type has an overall effect on yield.

### Water

Tests whether water level has an overall effect on yield.

### Fertilizer × Water

Tests whether the effect of fertilizer changes depending on water level.

### Predictive Analytics Connection

Interaction effects are important in predictive modelling.

A variable may have a small effect on its own but become highly informative when combined with another variable.

This idea later appears in:

- Interaction features
- Polynomial features
- Tree-based models

# Case Study 8: F-Test for Comparing Variances

## Business Problem

A manufacturer uses two machines.

Both machines may produce products with similar average dimensions, but the company also cares about consistency.

The question is:

> Is the variability of Machine A different from the variability of Machine B?

We compare:

$$
\sigma_A^2
$$

and:

$$
\sigma_B^2
$$

The F-statistic is:

$$
F=
\frac{s_A^2}{s_B^2}
$$

````python
machine_a = np.array([
    10.1, 10.2, 9.9, 10.0, 10.1,
    10.2, 10.0, 9.8, 10.1, 10.0
])

machine_b = np.array([
    9.4, 10.8, 9.2, 10.7, 9.5,
    10.6, 9.3, 10.9, 9.6, 10.5
])

variance_a = np.var(machine_a, ddof=1)
variance_b = np.var(machine_b, ddof=1)

f_statistic = variance_a / variance_b

df1 = len(machine_a) - 1
df2 = len(machine_b) - 1

cdf_value = stats.f.cdf(
    f_statistic,
    df1,
    df2
)

p_value = 2 * min(
    cdf_value,
    1 - cdf_value
)

print("Variance A:", round(variance_a, 4))
print("Variance B:", round(variance_b, 4))
print("F-Statistic:", round(f_statistic, 4))
print("P-Value:", round(p_value, 5))
````

````python
if p_value < alpha:
    print("Reject H0")
    print("The two machines have significantly different variances.")
else:
    print("Fail to Reject H0")
    print("There is not enough evidence that the machine variances differ.")
````

## Practical Interpretation

Average performance is not always enough.

Two processes can have similar averages but very different levels of consistency.

Variance testing can be useful in:

- Manufacturing quality control
- Financial risk
- Operational stability
- Model error analysis

The tests covered in this module are:

| Question | Test |
|---|---|
| One mean vs benchmark | One-Sample T-Test |
| Two independent means | Independent T-Test |
| Before vs after | Paired T-Test |
| Mean with known population standard deviation | Z-Test |
| Categorical vs categorical | Chi-Square Test |
| Three or more means | One-Way ANOVA |
| Two factors affecting a numerical outcome | Two-Way ANOVA |
| Comparing variances | F-Test |

The main purpose of statistical testing in predictive analytics is not simply to produce p-values.

It helps us ask better questions about data:

- Is this observed difference likely to be meaningful?
- Are these variables associated?
- Do groups behave differently?
- Has an intervention produced measurable change?
- Could a feature contain useful information about the target?

These questions help us understand the data before moving toward predictive modelling.

The next stage is to apply this reasoning to a real dataset, where data cleaning, exploratory analysis, statistical testing, and feature engineering become part of one continuous predictive analytics workflow.
