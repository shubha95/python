## 1. Why Probability Matters

Suppose an online store receives orders every day.

The business may want to answer questions such as:

- Will a visitor place an order?
- How many orders will arrive in the next hour?
- How much will a customer spend?
- How long will delivery take?

The exact outcome is not known beforehand. However, historical data can help us understand how likely different outcomes are.

This is where probability becomes useful.

For an event $A$, probability is written as:

$$
P(A)
$$

and its value lies between:

$$
0 \leq P(A) \leq 1
$$

A probability close to $0$ represents an unlikely event, while a probability close to $1$ represents a highly likely event.

```python
# A simple probability example

total_visitors = 1000
customers_who_purchased = 120

purchase_probability = customers_who_purchased / total_visitors

print("Probability of purchase:", purchase_probability)
print("Percentage:", purchase_probability * 100, "%")
```

### Observation

Based on historical data, the estimated probability that a visitor makes a purchase is:

$$
P(\text{Purchase}) = \frac{120}{1000} = 0.12
$$

This does not mean that exactly 12 out of every 100 future visitors must purchase. It means that, based on the observed data, a purchase occurs about 12% of the time.

Probability describes uncertainty. It does not remove it.

## 2. Random Variables

A random variable converts the outcome of a random process into a numerical value.

Consider three business questions:

| Situation | Random Variable |
|---|---|
| Whether a customer purchases | $X = 0$ or $1$ |
| Number of orders in an hour | $X = 0,1,2,3,\ldots$ |
| Customer order value | $X > 0$ |

The value of $X$ is not known before the event occurs.

Random variables are broadly classified into two types:

### Discrete Random Variable

A discrete random variable takes countable values.

Examples:

- Number of purchases
- Number of customer complaints
- Number of defective products
- Number of website clicks

### Continuous Random Variable

A continuous random variable can take any value within a range.

Examples:

- Delivery time
- Customer spending
- Product weight
- Temperature

```python
# Examples of discrete and continuous observations

orders_per_hour = [4, 7, 5, 9, 6]          # Discrete
delivery_time = [28.4, 31.7, 25.9, 36.2]  # Continuous

print("Orders per hour:", orders_per_hour)
print("Delivery times:", delivery_time)
```

## 3. Probability Distributions

A probability distribution describes how probabilities are distributed across the possible values of a random variable.

Instead of asking only:

> What value did we observe?

we begin asking:

> What values are possible, and how likely is each value?

The type of distribution depends on the type of random variable.

$$
\text{Discrete Random Variable}
\rightarrow
\text{Probability Mass Function (PMF)}
$$

$$
\text{Continuous Random Variable}
\rightarrow
\text{Probability Density Function (PDF)}
$$

Both discrete and continuous variables can also be studied using a:

$$
\text{Cumulative Distribution Function (CDF)}
$$

# Part A: Discrete Probability Distributions

## 4. Bernoulli Distribution

A Bernoulli experiment has only two possible outcomes.

Common examples include:

- Purchase or No Purchase
- Churn or No Churn
- Default or No Default
- Defective or Not Defective

We usually represent the outcomes as:

$$
X =
\begin{cases}
1, & \text{Success} \\
0, & \text{Failure}
\end{cases}
$$

If the probability of success is $p$, then:

$$
P(X=1)=p
$$

and:

$$
P(X=0)=1-p
$$

### Case: Will a Website Visitor Purchase?

Suppose the probability of a visitor making a purchase is $0.12$.

Then:

$$
P(X=1)=0.12
$$

$$
P(X=0)=0.88
$$

```python
from scipy.stats import bernoulli

p = 0.12

print("Probability of purchase:", bernoulli.pmf(1, p))
print("Probability of no purchase:", bernoulli.pmf(0, p))
```

### Practical Meaning

Bernoulli distributions appear frequently in predictive analytics because many prediction problems have two possible outcomes.

For example:

$$
\text{Will Customer Churn?}
\rightarrow
\{0,1\}
$$

$$
\text{Will Customer Default?}
\rightarrow
\{0,1\}
$$

A single Bernoulli trial describes one such event.

## 5. Binomial Distribution

Now suppose we observe many independent Bernoulli trials.

Instead of asking:

> Will one customer purchase?

we ask:

> How many customers will purchase out of 20 visitors?

If:

- There are $n$ independent trials
- Each trial has two possible outcomes
- The probability of success $p$ remains constant

then the number of successes follows a Binomial distribution.

$$
X \sim \text{Binomial}(n,p)
$$

The probability of exactly $k$ successes is:

$$
P(X=k)
=
\binom{n}{k}
p^k(1-p)^{n-k}
$$

### Case: Customer Conversion

Suppose:

$$
n=20
$$

and:

$$
p=0.12
$$

What is the probability that exactly 3 visitors make a purchase?

```python
from scipy.stats import binom

n = 20
p = 0.12
k = 3

probability = binom.pmf(k, n, p)

print("Probability of exactly 3 purchases:", round(probability, 4))
```

We can also ask a cumulative question:

> What is the probability that at most 3 visitors purchase?

This means:

$$
P(X \leq 3)
$$

```python
probability_at_most_3 = binom.cdf(3, n, p)

print("Probability of at most 3 purchases:", round(probability_at_most_3, 4))
```

### Bernoulli vs Binomial

The distinction is simple:

$$
\text{One Trial}
\rightarrow
\text{Bernoulli}
$$

$$
\text{Number of Successes in Multiple Trials}
\rightarrow
\text{Binomial}
$$

## 6. Poisson Distribution

The Poisson distribution is commonly used to model the number of times an event occurs within a fixed interval of time, area, distance, or space.

Examples:

- Customer calls per hour
- Orders per minute
- Server failures per month
- Accidents per day

If events occur at an average rate $\lambda$, then:

$$
X \sim \text{Poisson}(\lambda)
$$

The probability of observing exactly $k$ events is:

$$
P(X=k)
=
\frac{e^{-\lambda}\lambda^k}{k!}
$$

### Case: Customer Support Calls

A support team receives an average of 5 calls per hour.

What is the probability of receiving exactly 7 calls in the next hour?

```python
from scipy.stats import poisson

average_calls = 5
required_calls = 7

probability = poisson.pmf(required_calls, average_calls)

print("Probability of exactly 7 calls:", round(probability, 4))
```

Now consider:

> What is the probability of receiving 7 or fewer calls?

This is:

$$
P(X \leq 7)
$$

```python
probability = poisson.cdf(7, average_calls)

print("Probability of 7 or fewer calls:", round(probability, 4))
```

## 7. Probability Mass Function — PMF

A Probability Mass Function gives the probability that a discrete random variable takes an exact value.

For a discrete variable $X$:

$$
P(X=x)
$$

is meaningful.

For example:

$$
P(\text{Orders}=5)
$$

The Bernoulli, Binomial, and Poisson examples above all use PMFs when calculating the probability of an exact outcome.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

x = np.arange(0, 16)
pmf_values = poisson.pmf(x, mu=5)

plt.figure(figsize=(9, 4))
plt.bar(x, pmf_values)
plt.xlabel("Number of Calls")
plt.ylabel("Probability")
plt.title("PMF: Number of Support Calls per Hour")
plt.xticks(x)
plt.show()
```

### Observation

Each bar represents the probability of an exact count.

For example:

$$
P(X=5)
$$

is represented by the height of the bar at $X=5$.

Because the possible values are countable, separate probabilities can be assigned to individual outcomes.

# Part B: Continuous Probability Distributions

## 8. Uniform Distribution

In a continuous uniform distribution, all values within a specified interval have equal density.

Suppose a delivery arrives at any time between 20 and 40 minutes with equal likelihood.

We can represent this as:

$$
X \sim U(20,40)
$$

The probability density is:

$$
f(x)=\frac{1}{b-a}
$$

for:

$$
a \leq x \leq b
$$

### Case: Delivery Time

What is the probability that delivery takes between 25 and 30 minutes?

```python
from scipy.stats import uniform

a = 20
b = 40

probability = uniform.cdf(30, loc=a, scale=b-a) - uniform.cdf(25, loc=a, scale=b-a)

print("Probability of delivery between 25 and 30 minutes:", round(probability, 4))
```

For continuous variables, probability is associated with an interval rather than one exact point.

For example:

$$
P(25 \leq X \leq 30)
$$

This distinction becomes important when working with Probability Density Functions.

## 9. Normal Distribution

The Normal distribution is one of the most important distributions in statistics.

It is defined by two parameters:

- Mean: $\mu$
- Standard deviation: $\sigma$

We write:

$$
X \sim N(\mu,\sigma^2)
$$

The distribution is:

- Symmetric around the mean
- Bell-shaped
- Completely determined by $\mu$ and $\sigma$

Many real-world measurements can be approximately normal, although real data should never be assumed to be normal without checking.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 100
sigma = 15

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(9, 4))
plt.plot(x, y)
plt.axvline(mu, linestyle="--", label="Mean")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.legend()
plt.show()
```

### The Empirical Rule

For a Normal distribution, approximately:

$$
68\%
$$

of observations lie within:

$$
\mu \pm 1\sigma
$$

Approximately:

$$
95\%
$$

lie within:

$$
\mu \pm 2\sigma
$$

and approximately:

$$
99.7\%
$$

lie within:

$$
\mu \pm 3\sigma
$$

This gives us a practical way to understand how far an observation is from the typical range.

## 10. Standard Normal Distribution and Z-Score

A Normal random variable can be standardized using a Z-score.

$$
Z = \frac{X-\mu}{\sigma}
$$

A Z-score tells us how many standard deviations an observation is away from the mean.

### Case: Customer Spending

Suppose average customer spending is:

$$
\mu = 2000
$$

with standard deviation:

$$
\sigma = 400
$$

A customer spends:

$$
X = 2800
$$

The Z-score is:

```python
x = 2800
mu = 2000
sigma = 400

z = (x - mu) / sigma

print("Z-score:", z)
```

### Interpretation

The Z-score is:

$$
Z=2
$$

The customer's spending is therefore 2 standard deviations above the mean.

Standardization allows values measured on different scales to be compared using a common reference distribution.

The Standard Normal distribution has:

$$
\mu=0
$$

and:

$$
\sigma=1
$$

This idea will appear again when we study statistical inference and hypothesis testing.

## 11. Probability Density Function — PDF

A Probability Density Function describes the density of a continuous random variable.

For a continuous variable:

$$
P(X=x)=0
$$

This does not mean that the value $x$ is impossible.

It means that probability for a continuous variable is calculated over an interval.

For example:

$$
P(90 \leq X \leq 110)
$$

is represented by the area under the probability density curve between 90 and 110.

```python
mu = 100
sigma = 15

x = np.linspace(40, 160, 1000)
y = norm.pdf(x, mu, sigma)

lower = 90
upper = 110

plt.figure(figsize=(9, 4))
plt.plot(x, y)

x_fill = x[(x >= lower) & (x <= upper)]
y_fill = norm.pdf(x_fill, mu, sigma)

plt.fill_between(x_fill, y_fill, alpha=0.3)
plt.axvline(lower, linestyle="--")
plt.axvline(upper, linestyle="--")

plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Probability as Area Under the PDF")
plt.show()

probability = norm.cdf(upper, mu, sigma) - norm.cdf(lower, mu, sigma)

print("Probability between 90 and 110:", round(probability, 4))
```

### PMF vs PDF

| PMF | PDF |
|---|---|
| Used for discrete variables | Used for continuous variables |
| Probability of an exact value can be calculated | Probability is calculated over an interval |
| Example: $P(X=5)$ | Example: $P(90<X<110)$ |
| Bernoulli, Binomial, Poisson | Normal, Uniform |

The important distinction is the nature of the random variable.

## 12. Cumulative Distribution Function — CDF

The Cumulative Distribution Function answers the question:

> What is the probability that the random variable is less than or equal to a given value?

Mathematically:

$$
F(x)=P(X \leq x)
$$

The CDF can be used with both discrete and continuous distributions.

### Case: Delivery Time

Suppose delivery time is normally distributed with:

$$
\mu=30
$$

and:

$$
\sigma=5
$$

What is the probability that a delivery is completed within 35 minutes?

We need:

$$
P(X \leq 35)
$$

```python
from scipy.stats import norm

mu = 30
sigma = 5

probability = norm.cdf(35, mu, sigma)

print("Probability of delivery within 35 minutes:", round(probability, 4))
```

The CDF accumulates probability from the left side of the distribution up to the specified value.

It is particularly useful for questions containing phrases such as:

- Less than
- At most
- Within
- No more than

```python
x = np.linspace(10, 50, 500)
cdf_values = norm.cdf(x, mu, sigma)

plt.figure(figsize=(9, 4))
plt.plot(x, cdf_values)
plt.axvline(35, linestyle="--")
plt.axhline(norm.cdf(35, mu, sigma), linestyle="--")

plt.xlabel("Delivery Time")
plt.ylabel("Cumulative Probability")
plt.title("CDF of Delivery Time")
plt.show()
```

## 13. Expected Value

The expected value represents the long-run average value of a random variable.

For a discrete random variable:

$$
E(X)=\sum_x xP(X=x)
$$

The expected value should not always be interpreted as an outcome that must actually occur. It represents the average outcome expected over many repetitions.

### Case: Expected Number of Purchases

Suppose 20 visitors arrive and each has a purchase probability of $0.12$.

For a Binomial random variable:

$$
E(X)=np
$$

Therefore:

$$
E(X)=20 \times 0.12
$$

```python
n = 20
p = 0.12

expected_purchases = n * p

print("Expected number of purchases:", expected_purchases)
```

### Interpretation

The expected number of purchases is:

$$
E(X)=2.4
$$

This does not mean that exactly 2.4 customers can purchase.

It means that if we repeatedly observe groups of 20 similar visitors, the average number of purchases across those groups would approach 2.4.

Expected value connects probability distributions with long-run business expectations.

# 14. Bringing the Concepts Together

Consider an e-commerce business.

A visitor arrives on the website.

### Question 1: Will the visitor purchase?

This is a binary outcome.

$$
X \in \{0,1\}
$$

A **Bernoulli distribution** can describe the outcome.

---

### Question 2: How many of the next 20 visitors will purchase?

Now we count successes across multiple trials.

A **Binomial distribution** can be appropriate.

---

### Question 3: How many support calls will arrive in the next hour?

We count events occurring within a fixed interval.

A **Poisson distribution** may be appropriate.

---

### Question 4: How long will a delivery take?

Delivery time is continuous.

A continuous distribution such as the **Normal** or another suitable distribution may describe the variable.

---

The complete flow is:

$$
\text{Business Process}
\rightarrow
\text{Random Outcome}
\rightarrow
\text{Random Variable}
\rightarrow
\text{Distribution}
\rightarrow
\text{Probability}
$$

Probability distributions allow us to describe uncertainty.

Inferential statistics takes the next step:

$$
\text{Sample Data}
\rightarrow
\text{Learn About the Population}
$$

That transition begins with sampling distributions and the Central Limit Theorem.
