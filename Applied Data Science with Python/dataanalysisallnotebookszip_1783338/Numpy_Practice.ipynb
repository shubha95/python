# Question 1: Daily Website Visitors

A startup recorded website visitors for a week.

visitors = [120, 135, 150, 110, 180, 200, 175]

### Tasks
1. Convert the list into a NumPy array.
2. Display the array and its data type.
3. Find the total number of visitors.
4. Find the average number of visitors.
5. Find the highest and lowest visitor counts.

````python
import numpy as np
visitors = [120, 135, 150, 110, 180, 200, 175]

arr = np.array(visitors)
print("Array : ",arr)
print("Datatype: ",arr.dtype)
print("Total number of visitors: ",arr.sum())
print("Average number of visitors: ",arr.mean())
print("Highest number of visitors: ",arr.max())
print("Lowest number of visitors: ",arr.min())
````

### Tasks

2. Find the mean marks.
3. Find the median marks.
4. Find the standard deviation.
5. Count how many students scored above 80.


````python
marks = [78, 85, 90, 67, 88, 92, 75]
arr = np.array(marks)
print("Mean marks: ",arr.mean())
print("Median marks: ",np.median(arr))
print("Standard deviation: ",arr.std())
print("Number of students scoring above 80: ",(arr>80).sum())
````

# Question 4: Product Inventory Matrix

An electronics store has inventory counts for 12 products.

inventory = [120, 85, 95, 140, 110, 75, 60, 130, 100, 90, 150, 80]

### Tasks

1. Convert the list into a NumPy array.
2. Reshape the array into 3 rows and 4 columns.
3. Display the shape of the new array.
4. Display the first row.
5. Display the last column.
6. Find the total inventory available.

````python
inventory = [120, 85, 95, 140, 110, 75, 60, 130, 100, 90, 150, 80]
arr = np.array(inventory)
print("Array: ",arr)

new_arr = arr.reshape(3,4)
print("New array: ",new_arr)
print("Shape of new array: ",new_arr.shape)

print("First row: ",new_arr[0])
print("Last column: ",new_arr[:,-1])
print("Total inventory available: ",new_arr.sum())
````

# Question 5: Airline Passenger Analysis

An airline records passengers travelling over 4 days.

passengers = np.array([120, 135, 150, 180, 160, 145, 170, 190])

### Tasks

1. Display the first passenger count.
2. Display the last passenger count.
3. Display passenger counts from Day 3 to Day 6.
4. Display the first 5 values.
5. Display every alternate value.
6. Display the array in reverse order.
7. Find the highest passenger count using indexing.

````python
passengers = np.array([120, 135, 150, 180, 160, 145, 170, 190])
print("First passenger count: ",passengers[0])
print("Last passenger count: ",passengers[-1])
print("Passenger counts from Day 3 to Day 6: ",passengers[2:6])
print("First 5 values: ",passengers[:5])
print("Every alternate value: ",passengers[::2])
print("Array in reverse order: ",passengers[::-1])

# sort the data on ascending terms
passengers.sort()
print("Highest passenger count: ",passengers[-1])
````

# Question 6: Retail Store Sales Matrix


```
sales = np.array([
    [12000, 15000, 18000, 20000],
    [10000, 13000, 17000, 19000],
    [14000, 16000, 21000, 22000]
])
```



- Rows = Stores
- Columns = Months

### Tasks

1. Display sales of Store 1.
2. Display sales of Month 3 for all stores.
3. Display sales of Store 2 during Months 2 and 3.
4. Display the first two stores.
5. Display the last two months.
6. Display only the middle 2 columns.
7. Extract a sub-matrix containing:
   - Store 1 and Store 2
   - Month 2 and Month 3

````python
sales = np.array([
    [12000, 15000, 18000, 20000],
    [10000, 13000, 17000, 19000],
    [14000, 16000, 21000, 22000]
])
print("Sales of Store 1: ",sales[0,:])
print("Sales of Month 3 for all stores: ",sales[:,2])
print("Sales of Store 2 during Months 2 and 3: ",sales[1,1:3])
print("First two stores: ","\n",sales[:2,:])
print("Last two months: ","\n",sales[:,-2:])
print("Middle two columns: ","\n",sales[:,1:3])
print("Sub-matrix: ","\n",sales[:2,1:3])
````

# Question 7: Factory Production Dashboard


```
production = np.array([
    [100, 120, 140],
    [90, 110, 130],
    [150, 160, 180],
    [80, 100, 120]
])
```



- Rows = Machines
- Columns = Shifts

### Tasks

1. Calculate total production.
2. Calculate average production.
3. Find maximum production value.
4. Find minimum production value.
5. Calculate production by machine.
6. Calculate production by shift.
7. Find average production per machine.
8. Identify the highest-producing machine.

````python
production = np.array([
    [100, 120, 140],
    [90, 110, 130],
    [150, 160, 180],
    [80, 100, 120]
])
print("Total production: ",production.sum())
print("Average production: ",production.mean())
print("Maximum production value: ",production.max())
print("Minimum production value: ",production.min())

print("Production by machine: ",production.sum(axis=1))
print("Production by shift: ",production.sum(axis=0))
print("Average production per machine: ",production.mean(axis=1))
total_production_per_machine = production.sum(axis=1)
best_machine = np.argmax(total_production_per_machine)
print("Highest-producing machine: ",best_machine + 1)
````
