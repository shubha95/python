# 3. Probability

Probability is the study of randomness and uncertainty. It forms the foundation for statistical inference and is crucial for understanding many real-world phenomena.

We'll use the Titanic dataset to explore probability concepts. This dataset contains information about passengers on the Titanic, including whether they survived or not.

First, let's load and examine our dataset:

```python
# Load the Titanic dataset
titanic = sns.load_dataset("titanic")

print(titanic.head())
print("\nDataset Info:")
print(titanic.info())
```

## 3.1 Basic Probability Concepts

### Sample Space and Events

The sample space is the set of all possible outcomes of an experiment. An event is a subset of the sample space.

In our Titanic dataset:
- The sample space could be all passengers.
- An event could be "passenger survived" or "passenger was female".

Let's calculate some basic probabilities:

```python
total_passengers = len(titanic)
survivors = titanic['survived'].sum()
females = (titanic['sex'] == 'female').sum()

print(f"Total passengers: {total_passengers}")
print(f"Passengers who survived: {survivors}")
print(f"Passengers who were female: {females}")
```

```python
prob_survival = survivors / total_passengers
prob_female = females / total_passengers

print(f"Probability of survival: {prob_survival:.4f}")
print(f"Probability of being female: {prob_female:.4f}")

```

```python
# Visualize these probabilities
plt.figure(figsize=(10, 6))
sns.barplot(x=['Survival', 'Female'], y=[prob_survival, prob_female])
plt.title('Basic Probabilities in Titanic Dataset')
plt.ylabel('Probability')
plt.ylim(0, 1)
plt.show()
```

## 3.2 Conditional Probability

Conditional probability is the probability of an event occurring, given that another event has already occurred.

Let's calculate the probability of survival given that a passenger was female:

```python
female_survivors = titanic[(titanic['sex'] == 'female') & (titanic['survived'] == 1)].shape[0]
prob_survival_given_female = female_survivors / females

print(f"Probability of survival given female: {prob_survival_given_female:.4f}")

# Visualize conditional probabilities
survival_by_sex = titanic.groupby('sex')['survived'].mean()

plt.figure(figsize=(10, 6))
sns.barplot(x=survival_by_sex.index, y=survival_by_sex.values)
plt.title('Probability of Survival by Sex')
plt.ylabel('Probability of Survival')
plt.ylim(0, 1)
plt.show()
```

## 3.3 Independence

Two events are independent if the occurrence of one does not affect the probability of the other.

Let's check if survival was independent of passenger class:

```python
prob_first_class = (titanic['pclass'] == 1).mean()
prob_survival_first_class = titanic[titanic['pclass'] == 1]['survived'].mean()

print(f"P(Survival): {prob_survival:.4f}")
print(f"P(First Class): {prob_first_class:.4f}")
print(f"P(Survival | First Class): {prob_survival_first_class:.4f}")

```

```python
if abs(prob_survival - prob_survival_first_class) < 0.01:
    print("Survival and being in first class appear to be independent.")
else:
    print("Survival and being in first class do not appear to be independent.")

# Visualize survival probability by class
survival_by_class = titanic.groupby('pclass')['survived'].mean()

plt.figure(figsize=(10, 6))
sns.barplot(x=survival_by_class.index, y=survival_by_class.values)
plt.title('Probability of Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Probability of Survival')
plt.ylim(0, 1)
plt.show()
```

## 3.4 Bayes' Theorem

Bayes' Theorem is a way to update probabilities based on new evidence. It's expressed as:

**P(A|B) = (P(B|A) * P(A)) / P(B)**

Let's use Bayes' Theorem to calculate the probability that a passenger was in first class, given that they survived:

```python
P_survival = prob_survival
P_first_class = prob_first_class
P_survival_given_first_class = prob_survival_first_class

P_first_class_given_survival = (P_survival_given_first_class * P_first_class) / P_survival

print(f"P(First Class | Survival) = {P_first_class_given_survival:.4f}")

# Visualize the comparison
plt.figure(figsize=(10, 6))
sns.barplot(x=['P(First Class)', 'P(First Class | Survival)'],
            y=[P_first_class, P_first_class_given_survival])
plt.title('Probability of Being in First Class: Overall vs Given Survival')
plt.ylabel('Probability')
plt.ylim(0, 1)
plt.show()
```

## Binomial Distribution Sample Question

Let's consider a scenario where a company manufactures light bulbs. Historically, 10% of the light bulbs produced are defective. If we randomly select a batch of 50 light bulbs, what is the probability distribution of the number of defective bulbs in this batch?

We can simulate this process multiple times to observe the distribution of defective bulbs and then visualize it.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameters for the Binomial distribution
n = 50  # Number of trials (light bulbs in a batch)
p = 0.1 # Probability of success (a bulb being defective)

# Simulate 1000 batches of 50 light bulbs
num_simulations = 1000
defective_bulbs = np.random.binomial(n, p, num_simulations)

# Visualize the distribution of defective bulbs
plt.figure(figsize=(10, 6))
sns.histplot(defective_bulbs, bins=np.arange(-0.5, n + 1.5), kde=False, stat='count', discrete=True)
plt.title(f'Distribution of Defective Light Bulbs in {num_simulations} Batches (n={n}, p={p})')
plt.xlabel('Number of Defective Bulbs')
plt.ylabel('Frequency of Batches')
plt.xticks(np.arange(0, n + 1, 5)) # Set x-axis ticks at reasonable intervals
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Calculate mean and standard deviation of the simulated data
mean_defective = np.mean(defective_bulbs)
std_defective = np.std(defective_bulbs)

print(f"Average number of defective bulbs per batch: {mean_defective:.2f}")
print(f"Standard deviation of defective bulbs: {std_defective:.2f}")
```

## Poisson Distribution Sample Question

Let's consider a scenario where a call center receives an average of 7 calls per hour. We want to understand the probability distribution of the number of calls received in a given hour.

We can simulate this process multiple times to observe the distribution of calls per hour and then visualize it.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameter for the Poisson distribution
lambda_val = 7  # Average number of calls per hour

# Simulate the number of calls over 1000 hours
num_simulations = 1000
calls_per_hour = np.random.poisson(lambda_val, num_simulations)

# Visualize the distribution of calls per hour
plt.figure(figsize=(10, 6))
sns.histplot(calls_per_hour, bins=np.arange(-0.5, max(calls_per_hour) + 1.5), kde=False, stat='count', discrete=True)
plt.title(f'Distribution of Calls Per Hour in {num_simulations} Simulations (λ={lambda_val})')
plt.xlabel('Number of Calls Per Hour')
plt.ylabel('Frequency of Hours')
plt.xticks(np.arange(0, max(calls_per_hour) + 1, 1)) # Set x-axis ticks for each integer value
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Calculate mean and standard deviation of the simulated data
mean_calls = np.mean(calls_per_hour)
std_calls = np.std(calls_per_hour)

print(f"Average number of calls per hour: {mean_calls:.2f}")
print(f"Standard deviation of calls per hour: {std_calls:.2f}")
```

## Poisson Probability Mass Function (PMF)

The Probability Mass Function (PMF) for a Poisson distribution gives the probability that an event occurs exactly $k$ times in a fixed interval of time or space, given the average rate of occurrence $\lambda$ (lambda).

The formula is:

$P(X=k) = \frac{\lambda^k e^{-\lambda}}{k!}$

Where:
*   $P(X=k)$ is the probability of observing exactly $k$ events.
*   $\lambda$ (lambda) is the average number of events in the interval.
*   $k$ is the actual number of events (a non-negative integer).
*   $e$ is Euler's number (approximately 2.71828).
*   $k!$ is the factorial of $k$.

Let's calculate and visualize the PMF for our call center example, where $\lambda = 7$ calls per hour.

```python
from scipy.stats import poisson
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Parameters for the Poisson distribution from the previous example
lambda_val = 7  # Average number of calls per hour

# Define the range of k values (number of calls) to consider
# We'll go a bit beyond the mean to capture the distribution shape
k_values = np.arange(0, 20) # From 0 calls to 19 calls

# Calculate the PMF for each k value
poisson_pmf = [poisson.pmf(k, lambda_val) for k in k_values]

# Visualize the PMF with color coding
plt.figure(figsize=(12, 7))
sns.barplot(x=k_values, y=poisson_pmf, hue=k_values, palette='viridis', legend=False)
plt.title(f'Poisson Probability Mass Function (λ = {lambda_val})')
plt.xlabel('Number of Calls (k)')
plt.ylabel('Probability P(X=k)')
plt.xticks(k_values)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add probability values on top of bars for clarity
for i, prob in enumerate(poisson_pmf):
    plt.text(i, prob + 0.005, f'{prob:.3f}', ha='center', va='bottom', fontsize=9)

plt.show()
```

### Observation

The bar plot above visually represents the Poisson PMF for an average of $\lambda = 7$ calls per hour. Each bar shows the exact probability of receiving a specific number of calls ($k$) in an hour.

Key observations:
*   The distribution is skewed to the right, which is common for Poisson distributions when $\lambda$ is relatively small.
*   The highest probabilities are concentrated around the mean ($\lambda = 7$). This indicates that it's most probable to receive 6, 7, or 8 calls in an hour.
*   As you move further away from the mean, in either direction, the probabilities decrease, suggesting it's less likely to receive very few calls (e.g., 0-2) or a very large number of calls (e.g., 15 or more).

## Continous Probability Distributions

## Normal Distribution

The Normal Distribution, also known as the Gaussian distribution or bell curve, is one of the most important continuous probability distributions. It is characterized by two parameters: its mean (μ) and its standard deviation (σ).

Key characteristics:
*   **Symmetry:** The distribution is symmetric around its mean.
*   **Bell-shaped:** The graph has a distinctive bell shape.
*   **Mean, Median, Mode are Equal:** In a perfect normal distribution, all three measures of central tendency are at the same point.
*   **Parameters:** Defined by its mean (μ) and standard deviation (σ).

Many natural phenomena follow a normal distribution, such as heights, blood pressure, and measurement errors.

Let's generate a sample from a normal distribution and visualize it to see these characteristics.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameters for the Normal distribution
mean = 100    # Population mean (μ)
std_dev = 15  # Population standard deviation (σ)
num_samples = 10000 # A larger sample size for a smoother histogram

# Generate samples from a Normal distribution
normal_samples = np.random.normal(loc=mean, scale=std_dev, size=num_samples)

# Visualize the distribution with a histogram
plt.figure(figsize=(10, 6))
sns.histplot(normal_samples, bins=50, kde=True, color='skyblue', edgecolor='black')

# Add vertical lines for mean and standard deviation
plt.axvline(mean, color='red', linestyle='dashed', linewidth=2, label=f'Mean (μ={mean})')
plt.axvline(mean + std_dev, color='green', linestyle='dashed', linewidth=2, label=f'μ + σ ({mean + std_dev:.2f})')
plt.axvline(mean - std_dev, color='green', linestyle='dashed', linewidth=2, label=f'μ - σ ({mean - std_dev:.2f})')

plt.title(f'Histogram of {num_samples} Normal Samples (μ={mean}, σ={std_dev})')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Calculate mean and standard deviation of the simulated data
sim_mean = np.mean(normal_samples)
sim_std = np.std(normal_samples)

print(f"Simulated Mean: {sim_mean:.2f}")
print(f"Simulated Standard Deviation: {sim_std:.2f}")
```

## Normal Probability Density Function (PDF)

For continuous distributions like the Normal distribution, we use a Probability Density Function (PDF) instead of a PMF. The PDF describes the relative likelihood for this random variable to take on a given value. Unlike PMF, the value of the PDF at a given point is not a probability; rather, the probability of the variable falling within a certain range is found by integrating the PDF over that range.

The formula for the Normal Distribution's PDF is:

$f(x | \mu, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}$

Where:
*   $f(x)$ is the probability density at value $x$.
*   $\mu$ is the mean of the distribution.
*   $\sigma$ is the standard deviation of the distribution.
*   $\pi$ is approximately 3.14159.
*   $e$ is Euler's number (approximately 2.71828).

Let's calculate and visualize the PDF for our Normal distribution example (μ = 100, σ = 15).

```python
from scipy.stats import norm
import matplotlib.pyplot as plt
import numpy as np

# Parameters for the Normal distribution
mean = 100    # Population mean (μ)
std_dev = 15  # Population standard deviation (σ)

# Define the range of x values for the PDF plot
# Go a few standard deviations from the mean to cover most of the distribution
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 500)

# Calculate the PDF values for each x
pdf_values = norm.pdf(x, loc=mean, scale=std_dev)

# Visualize the PDF
plt.figure(figsize=(10, 6))
plt.plot(x, pdf_values, color='darkblue', linewidth=2, linestyle='-',marker = 'x')


# Add shading for illustrative purposes (e.g., probability within one std dev)
x_fill_lower = x[x >= (mean - std_dev)]
x_fill_upper = x_fill_lower[x_fill_lower <= (mean + std_dev)]
plt.fill_between(x_fill_upper, 0, norm.pdf(x_fill_upper, loc=mean, scale=std_dev), color='lightgreen', alpha=0.5, label='Within 1 Std Dev (68%)')

# Add vertical lines for mean and standard deviations
plt.axvline(mean, color='red', linestyle='--', label=f'Mean (μ={mean})')
plt.axvline(mean + std_dev, color='green', linestyle='--', label=f'μ + σ ({mean + std_dev})')
plt.axvline(mean - std_dev, color='green', linestyle='--', label=f'μ - σ ({mean - std_dev})')

plt.title(f'Normal Distribution Probability Density Function (PDF) (μ={mean}, σ={std_dev})')
plt.xlabel('Value (x)')
plt.ylabel('Probability Density f(x)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Observation

The plot above displays the Probability Density Function (PDF) of a Normal Distribution with a mean (μ) of 100 and a standard deviation (σ) of 15.

Key observations:
*   **Bell Shape:** The curve clearly exhibits the characteristic bell shape, peaking at the mean.
*   **Symmetry:** The curve is symmetric around the mean, indicating that values are equally likely to fall above or below the mean at equivalent distances.
*   **Spread:** The standard deviation dictates the spread of the curve. A larger standard deviation would result in a wider, flatter curve, while a smaller one would yield a narrower, taller curve.
*   **Area Under the Curve:** The total area under the entire curve is equal to 1, representing the sum of all probabilities. The shaded region (light green) illustrates the probability of a value falling within one standard deviation of the mean, which is approximately 68% for a normal distribution.

## Exponential Distribution

The Exponential Distribution is a continuous probability distribution that describes the time between events in a Poisson process, i.e., a process in which events occur continuously and independently at a constant average rate. It is often used to model the waiting time until the next event.

Key characteristics:
*   **Memoryless Property:** The probability of an event occurring in the future is independent of how much time has already passed.
*   **Single Parameter:** It is characterized by a single parameter, $\lambda$ (lambda), which is the rate parameter (the average number of events per unit of time).
*   **Decreasing Density:** The probability density is highest at zero and decreases exponentially as time increases.

Examples include the time until the next customer arrives, the lifetime of a device, or the time until the next phone call.

Let's generate samples from an Exponential distribution and visualize its shape.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameter for the Exponential distribution
lambda_rate = 0.5 # Rate parameter (e.g., average of 0.5 events per minute)
                   # This means, on average, an event occurs every 1/lambda = 2 minutes.
num_samples = 10000 # Number of samples to generate

# Generate samples from an Exponential distribution
# Note: numpy uses 'scale' parameter which is 1/lambda_rate
exponential_samples = np.random.exponential(scale=1/lambda_rate, size=num_samples)

# Visualize the distribution with a histogram
plt.figure(figsize=(10, 6))
sns.histplot(exponential_samples, bins=50, kde=True, color='purple', edgecolor='black')

plt.title(f'Histogram of {num_samples} Exponential Samples (λ={lambda_rate})')
plt.xlabel('Time (Units)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Calculate mean and standard deviation of the simulated data
sim_mean = np.mean(exponential_samples)
sim_std = np.std(exponential_samples)

print(f"Simulated Mean: {sim_mean:.2f}")
print(f"Simulated Standard Deviation: {sim_std:.2f}")
print(f"Theoretical Mean (1/λ): {1/lambda_rate:.2f}")
print(f"Theoretical Standard Deviation (1/λ): {1/lambda_rate:.2f}")
```

## Exponential Probability Density Function (PDF)

The PDF of the Exponential distribution gives the probability density for a continuous random variable $X$ representing the waiting time until the next event. The formula highlights its characteristic decreasing nature.

The formula for the Exponential Distribution's PDF is:

$f(x; \lambda) = \lambda e^{-\lambda x}$ for $x \ge 0$

Where:
*   $f(x)$ is the probability density at time $x$.
*   $\lambda$ is the rate parameter (average number of events per unit time).
*   $e$ is Euler's number (approximately 2.71828).

Let's calculate and visualize the PDF for our Exponential distribution example (λ = 0.5).

```python
from scipy.stats import expon
import matplotlib.pyplot as plt
import numpy as np

# Parameter for the Exponential distribution
lambda_rate = 0.5 # Rate parameter

# Define the range of x values for the PDF plot
# Go up to about 5 times the mean (1/lambda_rate) to cover most of the distribution
x = np.linspace(0, 5 * (1/lambda_rate), 500)

# Calculate the PDF values for each x
pdf_values = expon.pdf(x, scale=1/lambda_rate) # scale = 1/lambda

# Visualize the PDF
plt.figure(figsize=(10, 6))
plt.plot(x, pdf_values, color='darkgreen', linewidth=2, linestyle='-')

# Add vertical line for the mean
plt.axvline(1/lambda_rate, color='red', linestyle='--', label=f'Mean (1/λ = {1/lambda_rate:.2f})')

plt.title(f'Exponential Distribution Probability Density Function (PDF) (λ={lambda_rate})')
plt.xlabel('Time (x)')
plt.ylabel('Probability Density f(x)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Observation

The plot above displays the Probability Density Function (PDF) of an Exponential Distribution with a rate parameter (λ) of 0.5.

Key observations:
*   **Highest at Zero:** The probability density is highest at time $x=0$, indicating that shorter waiting times are more probable.
*   **Exponential Decay:** The curve demonstrates a continuous, exponential decrease as time increases. This signifies that the probability density for longer waiting times rapidly diminishes.
*   **Mean:** The red dashed line indicates the theoretical mean of the distribution, which is $1/\lambda = 1/0.5 = 2$. This represents the average waiting time between events.
*   **Shape:** The shape is characteristic of waiting times for events that occur at a constant average rate, where there's no 'memory' of past events affecting future probabilities.
