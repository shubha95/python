## Brief Report on the Statistical Analysis

**Dataset:** NSMES1988 (updated) — 4,406 records, 19 columns, no missing values in any column.

**1. Healthcare utilisation (visits, hospital, emergency, nvisits, ovisits)**
- `visits` (physician office visits) averages **5.77** but the median is only **4**, and the maximum is **89**. Skewness of **+3.34** confirms a strongly right-skewed distribution — most people visit the doctor a few times a year while a small group of heavy users pulls the mean up. For such a column the median describes a "typical" person better than the mean.
- `hospital` (mean 0.30) and `emergency` (mean 0.26) have a median and 75th percentile of **0**, i.e. more than three-quarters of the sample had no hospital stay and no emergency visit at all.

**2. Health status (chronic)**
- Number of chronic conditions averages **1.54** (median 1, max 8), so the average elderly respondent carries about one to two chronic conditions.

**3. Demographics after the Q3 transformation**
- `age` now reads in actual years: mean **74.0**, range **66 – 109**, mildly right-skewed (+0.89). The sample is entirely elderly (66+), as expected for a Medicare survey.
- `school` (years of education) averages **10.29** years, median 11, and is slightly left-skewed (−0.44) — most respondents completed high school or close to it.
- `income` now reads in dollars: mean **≈ $25,271**, median **≈ $16,982**, std **≈ $29,246**, range **−$10,125 to $548,351**. Skewness of **+5.96** shows a very heavy right tail — a few high-income respondents inflate the mean far above the median. The negative minimum is worth noting as a data-quality point (it represents a reported net loss rather than an error to be dropped blindly).

**4. Categorical variables**
- `health`: 3 categories, **average** dominates (3,509 of 4,406 ≈ 80%).
- `adl`: **normal** for 3,507 (≈ 80%); the rest have limited activities of daily living.
- `region`: 4 categories, **other** is the largest (1,614).
- `gender`: **female** majority (2,628 ≈ 60%); `married`: **yes** (2,406 ≈ 55%).
- `employed`: **no** for 3,951 (≈ 90%) — consistent with a retired population.
- `insurance`: **yes** for 3,421 (≈ 78%); `medicaid`: **no** for 4,004 (≈ 91%).

**5. Correlations**
- The strongest relationship is `hospital` ↔ `emergency` (**0.48**) — people admitted to hospital also use the emergency room more.
- `visits` correlates moderately with `chronic` (0.26), `hospital` (0.24) and `nvisits` (0.23): poorer health drives all forms of utilisation together.
- `school` ↔ `income` (**0.26**) is the only notable socio-economic link.
- `age` and `income` are essentially uncorrelated with `visits` (≈ 0.00), so utilisation in this elderly sample is explained by health condition rather than by age or income.

**Conclusion:** the dataframe is complete (no nulls) and describes an elderly, largely retired, mostly insured population in average self-reported health. The count-type utilisation columns and `income` are heavily right-skewed, so median/IQR should be preferred over mean/std for them, and any future modelling should use methods suited to skewed count data. The updated dataframe is saved below as `NSMES1988updated.csv`.

# Q5 Indicate which of the columns are not eligible for statistical analysis and indicate possible datatype changes, and report.
- Make changes to the recommended file from the previous step, export it as a new `.csv` file for possible future use (Optional).
- Prepare a brief report and enter it in the mark-up cells of the JupyterLab Notebook.

```python
# Step 1: Classify every column - is it eligible for statistical (numeric) analysis or not?
# "Eligible"     -> mean / std / correlation are meaningful
# "Not eligible" -> only counts and frequencies are meaningful (or the column carries no information)

review = {
    # column        nature                       eligible?              recommended dtype
    'Unnamed: 0': ('Row identifier (1..4406)',  'NOT eligible',        'drop the column'),
    'visits':     ('Discrete count',            'Eligible',            'int16'),
    'nvisits':    ('Discrete count',            'Eligible',            'int16'),
    'ovisits':    ('Discrete count',            'Eligible',            'int16'),
    'novisits':   ('Discrete count',            'Eligible',            'int16'),
    'emergency':  ('Discrete count',            'Eligible',            'int16'),
    'hospital':   ('Discrete count',            'Eligible',            'int16'),
    'health':     ('Ordinal category (3)',      'NOT eligible',        'ordered category'),
    'chronic':    ('Discrete count',            'Eligible',            'int16'),
    'adl':        ('Ordinal category (2)',      'NOT eligible',        'ordered category'),
    'region':     ('Nominal category (4)',      'NOT eligible',        'category'),
    'age':        ('Continuous, whole years',   'Eligible',            'int16'),
    'gender':     ('Nominal category (2)',      'NOT eligible',        'category'),
    'married':    ('Binary yes/no',             'NOT eligible as text','bool'),
    'school':     ('Discrete count (years)',    'Eligible',            'int16'),
    'income':     ('Continuous (dollars)',      'Eligible',            'float32'),
    'employed':   ('Binary yes/no',             'NOT eligible as text','bool'),
    'insurance':  ('Binary yes/no',             'NOT eligible as text','bool'),
    'medicaid':   ('Binary yes/no',             'NOT eligible as text','bool'),
}

eligibility = pd.DataFrame(
    [(c, str(df[c].dtype), n, e, r) for c, (n, e, r) in review.items()],
    columns=['column', 'current dtype', 'nature of data', 'statistical analysis', 'recommended dtype']
).set_index('column')

eligibility
```

```python
# Step 2: Why the count columns fit in int16 and age fits in int16 - check the actual ranges
# int8 holds -128..127, int16 holds -32768..32767
num_cols = ['visits', 'nvisits', 'ovisits', 'novisits', 'emergency',
            'hospital', 'chronic', 'age', 'school', 'income']

pd.DataFrame({'min': df[num_cols].min(),
              'max': df[num_cols].max(),
              'has decimals': [(df[c] % 1 != 0).any() for c in num_cols]})
```

```python
# Step 3: Apply the recommended changes on a copy of the dataframe
mem_before = df.memory_usage(deep=True).sum() / 1024      # KB, before any change

df_opt = df.copy()

# (a) drop the identifier column - it carries no information
df_opt = df_opt.drop(columns=['Unnamed: 0'])

# (b) counts and age -> int16
for col in ['visits', 'nvisits', 'ovisits', 'novisits', 'emergency',
            'hospital', 'chronic', 'school', 'age']:
    df_opt[col] = df_opt[col].astype('int16')

# (c) income -> float32 (still far more precision than the data needs)
df_opt['income'] = df_opt['income'].astype('float32')

# (d) ordinal text -> ORDERED category, so comparisons and sorting work
df_opt['health'] = pd.Categorical(df_opt['health'],
                                  categories=['poor', 'average', 'excellent'], ordered=True)
df_opt['adl'] = pd.Categorical(df_opt['adl'],
                               categories=['normal', 'limited'], ordered=True)

# (e) nominal text -> category
for col in ['region', 'gender']:
    df_opt[col] = df_opt[col].astype('category')

# (f) yes/no text -> bool (True means yes)
for col in ['married', 'employed', 'insurance', 'medicaid']:
    df_opt[col] = (df_opt[col] == 'yes')

df_opt.dtypes
```

```python
# Step 4: Compare the memory before and after the datatype changes
mem_after = df_opt.memory_usage(deep=True).sum() / 1024    # KB

print(f'Before : {mem_before:10.2f} KB   ({df.shape[1]} columns)')
print(f'After  : {mem_after:10.2f} KB   ({df_opt.shape[1]} columns)')
print(f'Saved  : {mem_before - mem_after:10.2f} KB  = {100*(1 - mem_after/mem_before):.1f} % reduction')

pd.DataFrame({'before (bytes)': df.memory_usage(deep=True),
              'after (bytes)': df_opt.memory_usage(deep=True)}).fillna(0).astype('int64')
```

```python
# Step 5: Confirm that nothing was lost - the statistics are the same after the conversion
comparison = pd.DataFrame({
    'mean before': df[['visits', 'chronic', 'age', 'school', 'income']].mean(),
    'mean after' : df_opt[['visits', 'chronic', 'age', 'school', 'income']].mean(),
}).round(2)
print(comparison)

# The columns that are NOT eligible for numeric statistics are analysed with frequencies
for col in ['health', 'adl', 'region', 'gender', 'married', 'employed', 'insurance', 'medicaid']:
    print(f'\n--- {col} ---')
    print(df_opt[col].value_counts())
```

```python
# Step 6 (optional part of the question): export the optimised dataframe as a NEW csv file
df_opt.to_csv('NSMES1988optimized.csv', index=False)

# IMPORTANT: a .csv file stores plain text only - it does NOT remember the dtypes.
# So when this file is read back, the dtype dictionary has to be supplied again.
dtype_map = {**{c: 'int16' for c in ['visits', 'nvisits', 'ovisits', 'novisits',
                                     'emergency', 'hospital', 'chronic', 'school', 'age']},
             'income': 'float32',
             **{c: 'category' for c in ['health', 'adl', 'region', 'gender']},
             **{c: 'bool' for c in ['married', 'employed', 'insurance', 'medicaid']}}

reloaded = pd.read_csv('NSMES1988optimized.csv', dtype=dtype_map)
print('Shape          :', reloaded.shape)
print('Memory reloaded: %.2f KB' % (reloaded.memory_usage(deep=True).sum() / 1024))
reloaded.head()
```
