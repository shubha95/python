## Basic Statistical Analysis & Report

The cells below analyse the **transformed** dataframe (where `age` was scaled by 10 and
`income` re-scaled), then save it as `NSMES1988updated.csv`:

1. `df.describe()` — summary statistics (count / mean / std / min / quartiles / max) for numeric columns.
2. Frequency counts for the categorical columns + a correlation matrix.
3. A short auto-generated text report of the key findings.
4. Export the dataframe to `NSMES1988updated.csv`.

```python
# Basic statistical summary of the NEW (transformed) dataframe
# describe() -> count, mean, std, min, quartiles, max for numeric columns
df.describe()
```

```python
# Categorical (object) columns: frequency of each category
for col in df.select_dtypes(exclude="number").columns:
    print(f"\n{col}:")
    print(df[col].value_counts())

# Correlation between numeric columns
df.select_dtypes(include="number").corr()
```

```python
# ---- Brief auto-generated report ----
n_rows, n_cols = df.shape
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(exclude="number").columns

print("=" * 60)
print("BRIEF STATISTICAL REPORT  -  NSMES1988 (transformed)")
print("=" * 60)
print(f"Records: {n_rows}   |   Columns: {n_cols}")
print(f"Numeric columns   : {list(num_cols)}")
print(f"Categorical columns: {list(cat_cols)}")
print(f"Missing values (total): {int(df.isna().sum().sum())}")
print("-" * 60)

# Key numeric highlights
print(f"Avg physician visits : {df['visits'].mean():.1f}  (max {df['visits'].max()})")
print(f"Avg age (years)      : {df['age'].mean():.1f}  "
      f"(range {df['age'].min():.0f}-{df['age'].max():.0f})")
print(f"Avg chronic conditions: {df['chronic'].mean():.2f}")
print(f"Median income        : {df['income'].median():,.0f}")
print("-" * 60)

# Categorical highlights
print("Self-perceived health:")
print(df['health'].value_counts().to_string())
print(f"Insured (private): {(df['insurance']=='yes').mean()*100:.1f}%")
print("=" * 60)
```

```python
# Save the new (transformed) dataframe for future use
df.to_csv("NSMES1988updated.csv", index=False)
print("Saved -> NSMES1988updated.csv")
print("Rows, Columns:", df.shape)
```
