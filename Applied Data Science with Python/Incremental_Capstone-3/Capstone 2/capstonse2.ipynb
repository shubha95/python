## Data Cleaning & Dtype Optimization Report

**Columns not eligible for standard statistical analysis**: `health`, `adl`, `region`, `gender`, `married`, `employed`, `insurance`, `medicaid` — all stored as text (`str`), so `describe()` cannot compute mean/std/quartiles on them; only frequency counts apply. `Unnamed: 0` is numeric but is a plain row index, not a real measurement, so it was dropped rather than analyzed.

**Changes applied**:
1. Dropped `Unnamed: 0` — reduced the dataframe from 19 to 18 columns.
2. Converted the 8 text/categorical columns to pandas `category` dtype (`health` as an ordered category: poor < average < excellent, since it has a natural ranking).

**Result — column count**: 19 → 18.

**Result — memory usage**: each converted column dropped from ~225,000–246,000 bytes (as `str`) to roughly 4,500–4,700 bytes (as `category`) — about a 98% reduction per column, and a combined saving of close to 1.8 MB across the dataset.

**Decision on keeping vs. dropping the categorical columns**: kept all 8 — they carry genuine demographic and health information needed for downstream analysis (e.g. insurance/medicaid coverage, health status, region). Unlike `Unnamed: 0`, these are meaningful variables, so the fix was a dtype conversion, not removal.

**Note on persistence**: CSV files don't store dtype metadata, so after re-loading `NSMES1988cleaned.csv`, these columns will read back as plain text and would need `.astype('category')` re-applied if the memory savings are needed again in a future session.

**Output file**: `NSMES1988cleaned.csv` — cleaned and dtype-optimized version, saved for future use.
