```python
# Step 1: bin age into groups (age column is in years/10, e.g. 6.9 = 69 yrs)
data['age_years'] = data['age'] * 10
bins = [60, 70, 80, 90, 100, 110]
labels = ['60-69', '70-79', '80-89', '90-99', '100+']
data['age_group'] = pd.cut(data['age_years'], bins=bins, labels=labels, right=False)
```

```python
# Step 2: count individuals per age group, split by gender
age_gender_dist = pd.crosstab(data['age_group'], data['gender'])
print(age_gender_dist)
```

# Capstone Session 3 — Data Analysis Report (NSMES1988 Healthcare Dataset)

## Dataset Overview
- **4406 records**, 18 columns, no missing values.
- **Numeric columns**: visits, nvisits, ovisits, novisits, emergency, hospital, chronic, age, school, income
- **Categorical columns**: health, adl, region, gender, married, employed, insurance, medicaid

## Categorical Variable Levels
| Column | Categories |
|---|---|
| health | average, poor, excellent |
| adl | normal, limited |
| region | other, midwest, northeast, west |
| gender | male, female |
| married | yes, no |
| employed | yes, no |
| insurance | yes, no |
| medicaid | no, yes |

## Visit Types — Overall Summary
Physician office visits (`visits`) dominate, averaging **5.77** per person (max 89). All other visit types (nvisits, ovisits, novisits, emergency, hospital) have medians of **0**, meaning most people never use those services — a small subset of high-utilizers drives the mean up (all show high std relative to mean → right-skewed/outlier-heavy).

## Visits by Gender
Females average **more visits** than males across most types:
- visits: female 6.01 vs male 5.42
- nvisits: female 1.73 vs male 1.45
- ovisits: male 0.85 vs female 0.68 (males slightly higher here)

## Visits by Marital Status
Unmarried individuals average slightly more visits (5.93) than married individuals (5.64), though nvisits/ovisits/hospital rates are close between groups.

## School (Education) by Gender
Nearly identical: female 10.26 years vs male 10.34 years — negligible gender gap in education.

## Income by Employment Status
Clear gap: employed individuals average **$3.98** (×10,000) vs **$2.36** for unemployed — confirms employment strongly drives income.

## Visits & Income by Insurance
Insured individuals have **both** higher visits (6.02 vs 4.91) and higher income (2.78 vs 1.66) than uninsured — insurance correlates with greater healthcare access.

## Visits & Income by Medicaid
Medicaid recipients have **higher visits** (6.71 vs 5.68) but **lower income** (1.14 vs 2.67) — consistent with Medicaid serving lower-income, higher-need populations.

## Age and Gender Distribution
| Age Group | Female | Male |
|---|---|---|
| 60-69 | 723 | 554 |
| 70-79 | 1369 | 924 |
| 80-89 | 475 | 265 |
| 90-99 | 59 | 34 |
| 100+ | 2 | 1 |

Females outnumber males in every age group, and the gap widens with age — consistent with longer average female life expectancy. Most individuals (both genders) fall in the 70-79 group.

## Health Status by Gender
| Health | Female | Male |
|---|---|---|
| average | 2093 | 1416 |
| excellent | 193 | 150 |
| poor | 342 | 212 |

Proportionally similar between genders (~80% average, ~7-8% excellent, ~12-13% poor) — no major gender disparity in self-reported health.

## Income Distribution by Gender
| Gender | Mean | Std | Min | Max |
|---|---|---|---|---|
| female | 2.25 | 2.72 | -1.01 | 54.84 |
| male | 2.94 | 3.16 | -1.01 | 54.84 |

Males average higher income than females (2.94 vs 2.25). Both genders share the same negative minimum (-1.01), a data-quality anomaly worth flagging.

## Regional Income Distribution
| Region | Mean | Max |
|---|---|---|
| west | 3.12 | 24.22 |
| northeast | 2.68 | 41.76 |
| midwest | 2.51 | 54.84 |
| other | 2.17 | 36.00 |

**West** has the highest average income; **"other"** the lowest — regional economic disparity is evident.

## Age-wise Income Analysis
| Age Group | Mean Income |
|---|---|
| 60-69 | 2.78 |
| 70-79 | 2.51 |
| 80-89 | 2.16 |
| 90-99 | 2.39 |
| 100+ | 1.55 |

Income **declines steadily with age** after 60-69, consistent with post-retirement income reduction, with a further drop at 100+.

## Key Observations
1. **Negative income values (-1.01, -0.82)** exist in the data — a cleaning issue to address before further modeling.
2. **Insurance and Medicaid status are the strongest differentiators** of both healthcare usage and income among all factors examined.
3. **Gender differences are modest** overall (visits, education, health) except in income, where males average notably higher.
4. **Age is inversely related to income**, and the population skews older (majority in 70-79 group) and female-heavy in older brackets.
5. **Region matters economically** — West residents earn most on average; "other" region residents earn least.
