```python
customer_List = ["John", "Jane", "Doe", "Smith","anish","shubham","satyam"]
 
```

```python
orderDetails = [
    ("John", "Laptop", 100, "Electronics"), 
    ("Jane", "Smartphone", 800, "Electronics"), 
    ("Doe", "Painting", 20, "Art"), 
    ("Smith", "glass", 60, "Home Decor"),
    ("anish", "book", 70, "Education"),
    ("shubham", "pencil", 10, "Education"),
    ("satyam", "rice", 20, "Grocery"),
     ("shubham", "pencil", 340, "Education"),
    ]
```

```python
orderProducts = {
    "John": ["Laptop", "Mouse"],
    "Jane": ["Smartphone", "Headphones"],
    "Doe": ["Painting", "Sculpture"],
    "Smith": ["Glass Vase", "Picture Frame"],
    "anish": ["Book", "Notebook"],
    "shubham": ["Pencil", "Eraser"],
    "satyam": ["Rice", "Pulses"]
}
```

```python
product_categories = {
    "Electronics": ["Laptop", "Smartphone", "Mouse", "Headphones"],
    "Art": ["Painting", "Sculpture"],
    "Home_Decor": ["Glass Vase", "Picture Frame"],
    "Education": ["Book", "Notebook", "Pencil", "Eraser"],
    "Grocery": ["Rice", "Pulses"],
}
```

```python
category = set()
for catagrey in product_categories:
    category.add(catagrey)
print(category)
```

```python
each_Customer_order = {}
for order in orderDetails:
    customer_name = order[0] 
    price = order[2]
    
    if customer_name not in each_Customer_order:
        each_Customer_order[customer_name] = price
        if customer_name in each_Customer_order:
            each_Customer_order[customer_name] += price
print("Each Customer Order:", each_Customer_order)
```

```python
classify_Customer = {}
for customer, total_price in each_Customer_order.items():
    if total_price > 100:
        classify_Customer[customer] = "high-value buyer"
    elif 50 <= total_price < 100:
        classify_Customer[customer] = "moderate buyer"
    else:
        classify_Customer[customer] = "low-value buyer"
print(str(classify_Customer))
```

```python
revanue_by_category = {}
for order in orderDetails:
    category = order[3] 
    price = order[2] 
    if category not in revanue_by_category:
        revanue_by_category[category] = price
    else:
        revanue_by_category[category] += price
print("Revenue by Category:", revanue_by_category)
```

```python
unique_products = set()
for order in orderDetails:
    product = order[1] 
    unique_products.add(product)
print("Unique Products:", unique_products)
```

```python
user_parchese_electronics = []
for order in orderDetails:
    if order[3] == "Electronics":
        user_parchese_electronics.append(order[0])
print("Users who purchased Electronics:", user_parchese_electronics)
```

```python
top_3_highest_spending = sorted(each_Customer_order.items(), key=lambda x: x[1], reverse=True)[:3]
print("Top 3 Highest Spending Customers:", top_3_highest_spending)
```

```python
print("Customer Summary Report")
print("=" * 60)
print(f"{'Customer':<15} {'Total Spending':<20} {'Classification':<20}")
print("=" * 60)

for customer in customer_List:
    if customer in each_Customer_order:
        spending = each_Customer_order[customer]
        classification = classify_Customer[customer]
        print(f"{customer:<15} ${spending:<19} {classification:<20}")

print("=" * 60)
```

```python
customers_electronics = set()
customers_education = set()

for order in orderDetails:
    if order[3] == "Electronics":
        customers_electronics.add(order[0])
    elif order[3] == "Education":
        customers_education.add(order[0])

common_customers = customers_electronics
print("Customers who bought both Electronics and Education:", common_customers)
```

```python
# Create a comprehensive customer order management system

# Display all customer orders with their products
print("Customer Order Management System")
print("=" * 70)
print(f"{'Customer':<15} {'Products':<30} {'Total Spending':<15}")
print("=" * 70)

for customer in customer_List:
    if customer in orderProducts and customer in each_Customer_order:
        products = ", ".join(orderProducts[customer])
        spending = each_Customer_order[customer]
        print(f"{customer:<15} {products:<30} ${spending:<14}")

print("=" * 70)

# Retrieve specific customer details using dictionary methods
print("\nCustomer Details Lookup:")
lookup_customer = "John"
if lookup_customer in each_Customer_order:
    print(f"{lookup_customer}'s Order Value: ${each_Customer_order[lookup_customer]}")
    print(f"{lookup_customer}'s Products: {', '.join(orderProducts[lookup_customer])}")
    print(f"{lookup_customer}'s Classification: {classify_Customer[lookup_customer]}")

# Update customer order (modify dictionary)
print("\nModifying Customer Records:")
print(f"Before: shubham's total spending = ${each_Customer_order['shubham']}")
each_Customer_order['shubham'] = 360  # Update spending
print(f"After: shubham's total spending = ${each_Customer_order['shubham']}")

# Summary statistics using dictionary methods
print("\nSummary Statistics:")
print(f"Total Customers: {len(customer_List)}")
print(f"Customers with Orders: {len(each_Customer_order)}")
print(f"Total Revenue: ${sum(each_Customer_order.values())}")
print(f"Average Spending per Customer: ${sum(each_Customer_order.values()) / len(each_Customer_order):.2f}")
```

```python
# Categorize customers by spending and analyze products by category

# Categorize customers based on spending levels
customer_spending_category = {}
for customer, spending in each_Customer_order.items():
    if spending > 100:
        category = "high-value"
    elif spending >= 50:
        category = "moderate"
    else:
        category = "low-value"
    
    if category not in customer_spending_category:
        customer_spending_category[category] = []
    customer_spending_category[category].append(customer)

print("Customers by Spending Category:")
for category, customers in customer_spending_category.items():
    print(f"{category}: {customers}")

# Find common and unique products across categories
print("\n" + "=" * 60)
print("Product Analysis by Category:")
print("=" * 60)

category_products = {}
for category_name, products in product_categories.items():
    category_products[category_name] = set(products)

# Find products common to Electronics and Education
common_products = category_products["Electronics"] & category_products["Education"]
print(f"Common products (Electronics & Education): {common_products if common_products else 'None'}")

# Find unique products in each category
for category_name, products_set in category_products.items():
    all_other_products = set().union(*[p for c, p in category_products.items() if c != category_name])
    unique = products_set - all_other_products
    print(f"Unique to {category_name}: {unique if unique else 'None'}")
```

```python
# Extract high-value customers and analyze trends

# High-value customers
high_value_customers = customer_spending_category['high-value']
print("High-Value Customers Analysis")
print("=" * 60)
print(f"High-Value Customers: {high_value_customers}")
print(f"Total Count: {len(high_value_customers)}")
print(f"Combined Spending: ${sum(each_Customer_order[c] for c in high_value_customers)}")
print(f"Average Spending: ${sum(each_Customer_order[c] for c in high_value_customers) / len(high_value_customers):.2f}")

# Most frequently purchased products
product_frequency = {}
for order in orderDetails:
    product = order[1]
    product_frequency[product] = product_frequency.get(product, 0) + 1

sorted_products = sorted(product_frequency.items(), key=lambda x: x[1], reverse=True)
print("\n" + "=" * 60)
print("Most Frequently Purchased Products:")
print("=" * 60)
for product, frequency in sorted_products:
    print(f"{product}: {frequency} purchase(s)")

# Category-wise sales trends
print("\n" + "=" * 60)
print("Category-Wise Sales Trends:")
print("=" * 60)
print(f"{'Category':<15} {'Revenue':<15} {'% of Total':<15}")
print("-" * 60)
total_revenue = sum(revanue_by_category.values())
for cat, revenue in sorted(revanue_by_category.items(), key=lambda x: x[1], reverse=True):
    percentage = (revenue / total_revenue) * 100
    print(f"{cat:<15} ${revenue:<14} {percentage:>6.1f}%")
print("-" * 60)
print(f"{'Total':<15} ${total_revenue:<14} {'100.0%':>6}")
```

```python
# Generate Comprehensive E-Commerce Analysis Report

print("\n" + "=" * 80)
print("COMPREHENSIVE E-COMMERCE ANALYSIS REPORT")
print("=" * 80)

# Section 1: Executive Summary
print("\n📊 EXECUTIVE SUMMARY")
print("-" * 80)
print(f"Total Customers: {len(customer_List)}")
print(f"Active Customers: {len(each_Customer_order)}")
print(f"Total Revenue: ${total_revenue:,}")
print(f"Average Order Value: ${total_revenue / len(each_Customer_order):.2f}")
print(f"Total Categories: {len(revanue_by_category)}")
print(f"Unique Products Sold: {len(unique_products)}")

# Section 2: Customer Classification Analysis
print("\n" + "=" * 80)
print("👥 CUSTOMER CLASSIFICATION ANALYSIS")
print("-" * 80)
high_value = customer_spending_category.get('high-value', [])
moderate = customer_spending_category.get('moderate', [])
low_value = customer_spending_category.get('low-value', [])

print(f"\nHigh-Value Buyers ({len(high_value)}): {', '.join(high_value)}")
print(f"  Combined Revenue: ${sum(each_Customer_order[c] for c in high_value)}")
print(f"  Average Spend: ${sum(each_Customer_order[c] for c in high_value) / len(high_value):.2f}")

print(f"\nLow-Value Buyers ({len(low_value)}): {', '.join(low_value)}")
print(f"  Combined Revenue: ${sum(each_Customer_order[c] for c in low_value)}")
print(f"  Average Spend: ${sum(each_Customer_order[c] for c in low_value) / len(low_value):.2f}")

# Section 3: Category Performance
print("\n" + "=" * 80)
print("📈 CATEGORY PERFORMANCE ANALYSIS")
print("-" * 80)
print(f"{'Category':<20} {'Revenue':<15} {'% Share':<15} {'Avg Order':<15}")
print("-" * 80)

for cat_name in sorted(revanue_by_category.keys()):
    revenue = revanue_by_category[cat_name]
    percentage = (revenue / total_revenue) * 100
    avg_order = revenue / len([o for o in orderDetails if o[3] == cat_name])
    print(f"{cat_name:<20} ${revenue:<14,} {percentage:>6.1f}%       ${avg_order:<14.2f}")

print("-" * 80)
print(f"{'TOTAL':<20} ${total_revenue:<14,} {'100.0%':>6}")

# Section 4: Product Insights
print("\n" + "=" * 80)
print("🛍️  PRODUCT INSIGHTS")
print("-" * 80)
print("\nTop Products by Frequency:")
for product, freq in sorted_products[:5]:
    print(f"  • {product}: {freq} order(s)")

# Section 5: Key Business Insights
print("\n" + "=" * 80)
print("💡 KEY BUSINESS INSIGHTS")
print("-" * 80)
print(f"1. Electronics dominates with {(revanue_by_category['Electronics']/total_revenue)*100:.1f}% of total revenue")
print(f"2. Top spender: Jane with ${each_Customer_order['Jane']} in orders")
print(f"3. {len(high_value)} out of {len(customer_List)} customers are high-value buyers")
print(f"4. 'Pencil' is the most frequently purchased item ({product_frequency['pencil']} purchases)")
print(f"5. Electronics & Education segments have no product overlap")

print("\n" + "=" * 80)
```
