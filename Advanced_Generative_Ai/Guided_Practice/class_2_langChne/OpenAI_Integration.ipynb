```python
from openai import OpenAI

client = OpenAI()

print("Client created successfully!")
```

```python
MODEL = "gpt-6-luna"  # Change this if your API account uses a different available model.

prompt = "Explain Generative AI to a first-year college student in 5 simple bullet points."

response = client.responses.create(
    model=MODEL,
    input=prompt
)

print(response.output_text)
```

```python
response = client.responses.create(
    model=MODEL,
    instructions=(
        "You are a friendly Python programming instructor. "
        "Use simple language, short paragraphs, and beginner-level examples."
    ),
    input="Explain the difference between a Python list and a tuple."
)

print(response.output_text)
```

```python
previous_id = None

print("Simple OpenAI chatbot. Type 'exit' to stop.")

while True:
    user_message = input("\\nYou: ").strip()

    if user_message.lower() in {"exit", "quit"}:
        print("Chatbot: Goodbye!")
        break

    if not user_message:
        print("Please enter a question.")
        continue

    # Include the instructions on every turn so the assistant keeps the same role.
    if previous_id is None:
        chat_response = client.responses.create(
            model=MODEL,
            instructions="You are a helpful beginner-friendly learning assistant.",
            input=user_message
        )
    else:
        chat_response = client.responses.create(
            model=MODEL,
            instructions="You are a helpful beginner-friendly learning assistant.",
            previous_response_id=previous_id,
            input=user_message
        )

    print("\\nAssistant:", chat_response.output_text)
    previous_id = chat_response.id
```
