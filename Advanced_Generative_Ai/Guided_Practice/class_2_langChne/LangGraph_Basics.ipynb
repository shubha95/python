## 2. Connect to Gemini API key

```python
from getpass import getpass
GOOGLE_API_KEY = getpass("Enter your Google API key : ")
```

```python
# initialize the model
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
                  google_api_key=GOOGLE_API_KEY,
                  model="gemini-3.5-flash"
)
```

## 3. Call the LLM

```python
# invoke - asking llm
# output - response
# prompt- query


response = llm.invoke("Whats the capital of Canada?")
print(response)
```

```python
# better ouput - formatted output
response.content[0]['text']
```

## 4. Create an Output Parser

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()

parsed_output = parser.invoke(response)
print(parsed_output)
```

## 5. Create a Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system",
         "You are an experienced Deep Learning expert."
         "Explain concept in not more than 100 words"),
        ("human"
         "Exaplin {question} with an example from {industry}.")
    ]
)
```

## 6. Connect everything with a Chain

```python
chain = prompt | llm | parser
```

```python
result = chain.invoke(
    {
        "question" : "What is a convolutional neural network?",
        "industry" : "Medical Diagonistics"
    }
)

print(result)
```
