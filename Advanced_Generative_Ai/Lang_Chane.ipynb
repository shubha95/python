## connection to gemini

```python
from getpass import getpass
GOOGLE_API_KEY = getpass("Enter your Google API Key: ")
```

```python
# initialize the model
from langchain_google_genai import ChatGoogleGenerativeAI
    
llm = ChatGoogleGenerativeAI(
    google_api_key=GOOGLE_API_KEY, 
    model="gemini-3.5-flash"
)
```

## Call the LLM

```python
# invoke - asking 11m
# output - response
# prompt- query

response = llm.invoke("Whats the capital of Canada?")
print(response)
```

```python
response.content[0]['text']
```

### create a prompt templete 

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an experienced Deep Learning expert. Explain concept in not more than 100 words"),
        ("human", "Exaplin {question} with an example from {industry}.")
    ]
)
```

```python
form_anssure = prompt.invoke({
    "question": "What is a neural network?",
    "industry": "Healthcare"})

print(form_anssure)
```

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
```

```python
paresed_output = parser.invoke(response)
print(paresed_output)
```

### 6. Connect everything with a Chain

```python
chain = prompt| llm | parser

```

```python
result = chain.invoke(
    {
        "question" : "What is a convolutional neural network?",
        "industry" : "Medical Diagonistics"
    }
)

print(result)
```
