# Functions: Generators

Generators produce values one at a time instead of building all values at once.

This is useful when:
- data is large
- memory use matters
- values are processed in a stream


# return vs yield Statement

- `return` sends a value back and ends the function.
- `yield` pauses the function and produces one value at a time.

Generators use `yield`.


````python
# Example 1: simple generator
def number_stream():
    yield 1
    yield 2
    yield 3

gen = number_stream()
print(next(gen))
print(next(gen))
print(next(gen))
````

````python
# Example 2: generator with a loop
def even_numbers(limit):
    for num in range(0, limit + 1, 2):
        yield num

for value in even_numbers(10):
    print(value)
````

````python
# Example 3: analytics-style generator
def monthly_sales():
    yield 1200
    yield 1500
    yield 1800

for sale in monthly_sales():
    print("Sale:", sale)
````

````python
# Example 4: generator for row numbers
def row_numbers():
    for i in range(1, 4):
        yield i

rows = row_numbers()
print(next(rows))
print(next(rows))
````

````python
# Example 5: generator for streaming data
def stream_scores():
    for score in [78, 82, 91]:
        yield score

for score in stream_scores():
    print("Score:", score)
````

# Demo: Implementing Generators, Lambda, and Built-in Functions

Case study: analytics summary processing.

We will:
- generate monthly values using a generator
- clean labels using lambda
- calculate totals using built-in functional tools


````python
from functools import reduce

# Generator for monthly values
def sales_stream():
    # Simulate a stream of monthly sales
    yield 1200
    yield 1450
    yield 1600
    yield 2100

# Lambda for cleaning a text label
clean_label = lambda text: text.strip().title()

# Use generator output in a list for functional processing
sales_values = list(sales_stream())

print("Clean Label:", clean_label("  monthly sales report  "))

# map: add a fixed uplift to each value
uplifted_sales = list(map(lambda x: x + 100, sales_values))
print("Uplifted Sales:", uplifted_sales)

# filter: keep only strong values
strong_sales = list(filter(lambda x: x >= 1500, sales_values))
print("Strong Sales:", strong_sales)

# reduce: calculate total sales
total_sales = reduce(lambda x, y: x + y, sales_values)
print("Total Sales:", total_sales)
````

# Demo: Implementing a Class with Attributes and Methods

Case study: an analytics student performance tracker.

The class will store student details and calculate a basic status.


````python
# Case study: student performance tracker
class StudentAnalytics:
    # Class attribute shared by all student records
    school = "Data and AI Academy"

    def __init__(self, name, marks):
        # Data attributes belonging to each object
        self.name = name
        self.marks = marks

    def average(self):
        # For demo purposes, average is same as marks here
        return self.marks

    def status(self):
        # Return a simple performance label
        if self.marks >= 90:
            return "Excellent"
        elif self.marks >= 70:
            return "Good"
        return "Needs Improvement"

student1 = StudentAnalytics("Asha", 92)
student2 = StudentAnalytics("Rahul", 78)

print(student1.name, student1.average(), student1.status(), student1.school)
print(student2.name, student2.average(), student2.status(), student2.school)
````

# Inheritance

Inheritance allows one class to use properties and methods from another class.

This is useful when we want to build specialized objects from a common base.


# Types of Inheritance

Common types include:
- single inheritance
- multiple inheritance
- multilevel inheritance
- hierarchical inheritance


````python
# Example 1: single inheritance
class BaseReport:
    def show(self):
        print("Base report")

class SalesReport(BaseReport):
    pass

r = SalesReport()
r.show()
````

````python
# Example 2: parent and child behavior
class Employee:
    def __init__(self, name):
        self.name = name

class Analyst(Employee):
    pass

a = Analyst("Sara")
print(a.name)
````

````python
# Example 3: inherited method
class Dataset:
    def load(self):
        print("Loading dataset")

class SalesDataset(Dataset):
    pass

sd = SalesDataset()
sd.load()
````

````python
# Example 4: override in child
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

sq = Square(5)
print(sq.area())
````

````python
# Example 5: analytics inheritance idea
class Report:
    def summary(self):
        print("Generic report summary")

class MonthlyReport(Report):

    def summary(self):
       print("Monthly analytics summary")

mr = MonthlyReport()
mr.summary()
````

# Demo: Implementing Inheritance

Case study: general report class and a specialized sales report class.


````python
# Case study: report inheritance
class Report:
    def __init__(self, title):
        self.title = title

    def display(self):
        print("Report Title:", self.title)

class SalesReport(Report):
    def __init__(self, title, revenue):
        super().__init__(title)
        self.revenue = revenue

    def display(self):
        # Extend the parent behavior
        super().display()
        print("Revenue:", self.revenue)

report = SalesReport("June Sales Report", 185000)
report.display()
````

# Abstraction

Abstraction means hiding internal details and showing only the essential features.

It helps users focus on what something does, not how it is implemented internally.


````python
# Example 1: method overriding
class Bird:
    def speak(self):
        return "Sound"

class Parrot(Bird):
    def speak(self):
        return "Squawk"

print(Bird().speak())
print(Parrot().speak())
````

````python
# Example 2: different objects, same method name
class Laptop:
    def start(self):
        print("Starting laptop")

class Mobile:
    def start(self):
        print("Starting mobile")

for device in [Laptop(), Mobile()]:
    device.start()
````

````python
# Example 3: simple abstraction idea
class Dashboard:
    def show(self):
        print("Showing summary")

class SalesDashboard(Dashboard):
    pass

d = SalesDashboard()
d.show()
````

````python
# Example 4: common interface style
class Alert:
    def send(self):
        print("Sending generic alert")

class EmailAlert(Alert):
    def send(self):
        print("Sending email alert")

print(Alert().send())
print(EmailAlert().send())
````

````python
# Example 5: analytics object behavior
class Metric:
    def display(self):
        print("Generic metric")

class RevenueMetric(Metric):
    def display(self):
        print("Revenue metric")

Metric().display()
RevenueMetric().display()
````

# Demo: Implementing File Handling in Python

Case study: analyzing `sales.txt` and `titanic.csv`.

We will:
- read sales values from a text file
- read passenger data from a CSV file
- calculate simple analytics summaries


````python
import csv

# ---- Sales analysis from sales.txt ----
sales_values = []

with open("sales.txt", "r", encoding="utf-8") as file:
    for line in file:
        # Each line looks like: January,1200
        month, amount = line.strip().split(",")
        sales_values.append(int(amount))

print("Sales Values:", sales_values)
print("Total Sales:", sum(sales_values))
print("Average Sales:", sum(sales_values) / len(sales_values))

# ---- Titanic analysis from titanic.csv ----
survived_count = 0
total_passengers = 0
first_class_count = 0

with open("titanic.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        total_passengers += 1

        # Convert values from text to numbers when needed
        survived = int(row["Survived"])
        pclass = int(row["Pclass"])

        if survived == 1:
            survived_count += 1

        if pclass == 1:
            first_class_count += 1

print("Total Passengers:", total_passengers)
print("Survived Passengers:", survived_count)
print("First Class Passengers:", first_class_count)
````

# 4. Error Handling in Python

Programs can fail because of:
- invalid input
- missing files
- wrong data types
- division by zero

Error handling helps the program remain stable instead of crashing immediately.


# Introduction to Error Handling

Python uses `try` and `except` blocks to handle exceptions.

Additional blocks:
- `else`
- `finally`


````python

# Example 1: catching a division error
num1 = int(input("Enter the numerator: "))
num2 = int(input("Enter the denominator:"))

try:
  result = num1 / num2
  print("Result:", result)
except ZeroDivisionError:
  print("Cannot divide by zero")
````

````python
0# Example 2: handling invalid input
try:
    value = int("abc")
except ValueError:
    print("Invalid number format")
````

````python
# Example 3: safe file open
try:
    with open("missing_file.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found")
````

````python
# Example 4: using else
try:
    number = int("25")
except ValueError: # treat it like if
    print("Conversion failed")
else: # else
    print("Converted successfully:", number)
````

````python
# Example 5: using finally
try:
    result = 10/5
    print(result)
    print("Trying a risky operation")
except Exception:
    print("Something went wrong")
finally:
    print("This block always runs")
````

# Demo: Implementing Error Handling in Python

Case study: reading sales data where some rows may be invalid.

We will:
- read a file
- safely convert values
- skip bad rows
- report the issue clearly


````python
# Create a small file with one invalid row for the demo
invalid_sales_text = (
    "January,1200\n"
    "February,1500\n"
    "March,invalid_value\n"
    "April,2100\n"
)

with open("sales_with_error.txt", "w", encoding="utf-8") as file:
    file.write(invalid_sales_text)

valid_sales = []
invalid_rows = []

with open("sales_with_error.txt", "r", encoding="utf-8") as file:
    for line in file:
        try:
            month, amount = line.strip().split(",")
            amount = int(amount)
            valid_sales.append(amount)
        except ValueError:
            # If conversion fails, keep track of the bad row
            invalid_rows.append(line.strip())

print("Valid Sales:", valid_sales)
print("Total Valid Sales:", sum(valid_sales))
print("Invalid Rows:", invalid_rows)
````
