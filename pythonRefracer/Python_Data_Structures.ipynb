# Challenge Question

An online food delivery platform charges:

- Food bill = 850
- Delivery fee = 60
- Coupon discount = 100

Calculate the final payable amount.

Then check:

- Is the payable amount less than 1000?

```python
food_bill = int(input("Enter food bill amount"))
delivery_fee = int(input("Enter delivery fee"))
coupon_discount = int(input("Enter coupon discount"))

final_amount = food_bill + delivery_fee - coupon_discount

print("Total Bill: ", final_amount)
if final_amount < 1000:
  print("Amount is less than 1000")
else:
  print("Amount is greater than 1000")
```
