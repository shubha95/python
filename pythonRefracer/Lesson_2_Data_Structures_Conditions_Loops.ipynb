# Lists

Lists are ordered and mutable collections.

Lists are useful when:
- order matters
- data changes frequently
- duplicate values may exist

Analytics teams often use lists to store:
- sales values
- customer IDs
- KPI scores
- category names


```python
# Creating a list of sales values
sales = [1200, 1500, 900, 1800]
print(sales)
```

```python
# Mixed data list
record = ["Asha", 92, True]
print(record)
```

```python
# Positive indexing
courses = ["Python","SQL","Power BI"]
print(courses[0])
print(courses[2])
```

```python
# Negative indexing
months = ["Jan","Feb","Mar","Apr"]
print(months[-1])
```

```python
# Slicing example
sales = [100,200,300,400,500]
print(sales[1:4])
```

```python
# append adds a new value
sales=[1200,1400]
sales.append(1600)
print(sales)
```

```python
# pop removes a value
sales=[1200,1400,1600]
sales.pop()
print(sales)
```

```python
# extend combines lists
team1=["Python","SQL"]
team2=["Excel","Power BI"]
team1.extend(team2)
print(team1)
```

```python
# insert places value at index
cities=["Delhi","Mumbai"]
cities.insert(1,"Bengaluru")
print(cities)
```

```python
# remove deletes specific item
cities=["Delhi","Mumbai","Delhi"]
cities.remove("Delhi")
print(cities)
```

```python
# sort arranges values
scores=[88,45,92,76]
scores.sort()
print(scores)
```

```python
# reverse reverses order
scores=[88,45,92]
scores.reverse()
print(scores)
```

```python
# len gives total count
customers=["A","B","C"]
print(len(customers))
```

# Tuples

Tuples are ordered but immutable collections.

Once created, tuple values cannot be changed.

# Dictionaries

Dictionaries store information in key-value pairs.

Useful for:
- customer profiles

# Sets

```python
# Membership check
print("Python" in skills)
```

```python
# Intersection
print(team1.intersection(team2))
```

```python
# Difference
print(team1.difference(team2))
```

# Conditional Statements

Conditions help programs make decisions.

Real-world analytics systems use conditions for:
- segmentation
- eligibility checks
- alerts
- fraud detection


```python
# if statement
sales=12000
if sales>10000:
    print("Target Achieved")
```

```python
# if-else
attendance=68
if attendance>=75:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
# grading using elif
score=82
if score>=90:
    print("A")
elif score>=75:
    print("B")
else:
    print("C")
```

```python
# Fraud flag example
amount=90000
if amount>50000:
    print("Review Transaction")
```

```python
# Customer segmentation
purchase=6500
if purchase>5000:
    print("Premium Customer")
else:
    print("Regular Customer")
```

```python
# Nested if
age=25
income=40000
if age>=21:
    if income>30000:
        print("Eligible")
```

```python
# Login example
password="python123"
if password=="python123":
    print("Login Successful")
```

```python
# match-case example
option=2
match option:
    case 1:
        print("Balance")
    case 2:
        print("Withdrawal")
    case _:
        print("Invalid")
```

# Loops

Loops repeat tasks automatically.

Analytics workflows often repeat operations on:
- rows
- customers
- transactions
- KPI lists


```python
# for loop on sales values
sales=[1200,1500,1700]
for s in sales:
    # print each sales value
    print(s)
```

```python
# range loop
for i in range(5):
    print(i)
```

```python
# Attendance example
students=["Ali","Sara","Rahul"]
for s in students:
    print("Present:",s)
```

```python
# while loop
count=1
while count<=3:
    print(count)
    count+=1
```

```python
# break example
for num in range(10):
    if num==5:
        break
    print(num)
```

```python
# continue example
for num in range(5):
    if num==2:
        continue
    print(num)
```

```python
# Nested loop
for i in range(2):
    for j in range(2):
        print(i,j)
```

```python
# KPI processing
scores=[75,88,92]
for score in scores:
    print("KPI:",score)
```

# Problem 1 — Student Analytics Dashboard

Store student marks in a list.

Use loops and conditions to:
- print marks
- identify students scoring above 80
- count total students


```python
# example solution

# Student names
students = ["Ali", "Sara", "Rahul", "Fatima", "John"]

# Student marks
marks = [78, 92, 85, 67, 95]

# Count total students using len()
total_students = len(students)

print("Student Marks Report")
print("---------------------")

# Loop through both lists together
for i in range(total_students):

    # Print student name and marks
    print(students[i], ":", marks[i])

    # Check students scoring above 80
    if marks[i] > 80:
        print("High Performer")

print("---------------------")
print("Total Students:", total_students)
```

# Solution Explanation

This program uses three important concepts together:

1. Lists
- students stores names
- marks stores scores

2. Loops
- for loop is used to go through every student record

3. Conditions
- if statement checks whether marks are above 80

The len() function counts the total number of students.

This is a simple example of how analytics dashboards process and evaluate data automatically.
